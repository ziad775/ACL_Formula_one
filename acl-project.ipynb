## 0.5 Load raw tables (immutable)
The raw tables are loaded **as-is**: the `\N` sentinel is deliberately kept as a string here,
so the raw EDA in Section 1 can measure how much of it exists before we clean it.

To guarantee the raw data is never modified, each table gets a fingerprint (hash) at load time.
`get_table()` always returns a copy, and `assert_raw_unchanged()` can be called at any point to prove the raw data is intact.

```python
raw = {t: pd.read_csv(DATA_DIR / f"{t}.csv", keep_default_na=False, na_values=[""])
       for t in TABLES}

def fingerprint(df: pd.DataFrame) -> str:
    return hashlib.md5(pd.util.hash_pandas_object(df, index=True).values.tobytes()).hexdigest()

RAW_FINGERPRINTS = {t: fingerprint(df) for t, df in raw.items()}

def get_table(name: str) -> pd.DataFrame:
    # Always work on a copy so the raw tables stay untouched.
    return raw[name].copy()

def assert_raw_unchanged() -> None:
    changed = [t for t, df in raw.items() if fingerprint(df) != RAW_FINGERPRINTS[t]]
    assert not changed, f"Raw tables were modified: {changed}"
    print("Raw tables unchanged ✔")

assert_raw_unchanged()
```

```python
# --- Reusable EDA helpers -------------------------------------------------------
# Defined once and used both before cleaning (Section 1) and after it (Section 3),
# so the before/after comparison uses exactly the same measurements.

def to_num(s):
    # Temporary numeric view of a column, for measuring only (does not clean anything).
    return pd.to_numeric(s, errors="coerce")

def is_missing(s):
    # A cell counts as missing if it is NaN/empty OR the literal sentinel "\N".
    return s.isna() | (s.astype(str) == SENTINEL)

def missing_profile(tables):
    # One row per (table, column): number and % of missing cells.
    rows = []
    for t, df in tables.items():
        for c in df.columns:
            m = is_missing(df[c])
            rows.append({"table": t, "column": c, "rows": len(df),
                         "missing": int(m.sum()), "missing_%": 100 * m.mean()})
    return pd.DataFrame(rows)

def with_year(df, races_df):
    # Attaches the season (and race name) to any table that has a raceId column.
    return df.merge(races_df[["raceId", "year", "name"]], on="raceId", how="left")
```

## 1.2 Is missingness random, or tied to eras?
If a column's missing share changes with the season, the value was simply **not recorded** in older eras.
If it is roughly constant, it depends on the **race outcome** (e.g. no finishing position for a retirement).

```python
# --- Missing % per season for selected results / races columns -----------------
def missing_by_season(df, col):
    return df.groupby("year")[col].apply(lambda s: is_missing(s).mean() * 100)

races_r = get_table("races")
season_missing = pd.DataFrame({
    "results.position":       missing_by_season(res_raw, "position"),
    "results.milliseconds":   missing_by_season(res_raw, "milliseconds"),
    "results.fastestLapTime": missing_by_season(res_raw, "fastestLapTime"),
    "races.time":             missing_by_season(races_r, "time"),
    "races.fp1_date":         missing_by_season(races_r, "fp1_date"),
})
season_missing.iloc[::10].round(1)   # every 10th season, as a quick look
```

**Finding:** there are three different *kinds* of missingness:

| Kind | Columns | Meaning |
|---|---|---|
| Outcome-based | `results.position`, `results.milliseconds` | Driver was not classified / did not finish. Roughly stable across eras. |
| Era-based | `results.fastestLapTime`, `races.time`, `races.fp1_date` (and other session columns) | Not recorded in older seasons: 100 % missing until a certain year, then filled. |

## 1.3 Missing rows, not just missing cells
`lap_times` and `pit_stops` showed **0 %** sentinels in cell 0.6, yet they don't cover every race.
Their missingness is **absent rows**, which cell counts cannot see. We measure it as: for each season, the share of
`results` rows that have at least one matching row in the other table.

```python
# --- Coverage of qualifying / lap_times / pit_stops per season --------------------
def coverage_by_season(base, other, keys=("raceId", "driverId")):
    # % of base rows (per season) that have a matching row in `other`.
    keys = list(keys)
    hits = other[keys].drop_duplicates().assign(_hit=True)
    has = base[keys + ["year"]].merge(hits, on=keys, how="left")["_hit"].notna()
    return has.groupby(base["year"].values).mean() * 100

coverage = pd.DataFrame({
    "qualifying": coverage_by_season(res_raw, raw["qualifying"]),
    "lap_times":  coverage_by_season(res_raw, raw["lap_times"]),
    "pit_stops":  coverage_by_season(res_raw, raw["pit_stops"]),
})
first_year = coverage.apply(lambda s: s[s > 0].index.min())
print("First season with any coverage:")
print(first_year.to_string())
```

```python
# --- Plot: coverage per season ------------------------------------------------------
fig, ax = plt.subplots(figsize=(11, 4.5))
coverage.plot(ax=ax)
ax.set_ylim(0, 105)
ax.set_xlabel("Season")
ax.set_ylabel("Results rows with a match (%)")
ax.set_title("Coverage drift: detailed tables only exist for recent decades")
ax.legend(frameon=False)
plt.tight_layout()
plt.show()
```

**Finding:** `qualifying`, `lap_times` and `pit_stops` have **zero** coverage before the first seasons printed above.
In earlier seasons the starting order is only known from `results.grid`. Coverage is also **not complete after it starts**:
qualifying has a deep dip in the late 1990s / early 2000s (many races in those seasons have no qualifying rows at all;
read the exact seasons from the plot), and only becomes near-complete from the early 2000s.

**Decision:** this is not something cleaning can fix (the data does not exist). It is handed over to the
features/modeling owner: qualifying-based features only exist from the mid-1990s, so they must either fall back to
`grid`, use a missing-indicator, or restrict the training window. This is the *qualifying-coverage drift* the brief mentions.

## 1.4 Are columns stored as the right type?
Every text column is reduced to the *shape* of its values (each digit → `9`), e.g. `"1:27.452"` → `9:99.999`.
The shapes tell us which columns are dates, clock times, durations, or numbers stored as text.

```python
# --- Value shapes of every text column -------------------------------------------------
def value_shapes(s):
    v = s[~is_missing(s)].astype(str)
    return v.str.replace(r"\d", "9", regex=True).value_counts()

rows = []
for t, df in raw.items():
    for c in df.columns:
        if pd.api.types.is_numeric_dtype(df[c]):
            continue                               # already numeric
        shapes = value_shapes(df[c])
        rows.append({"table": t, "column": c, "dtype": str(df[c].dtype),
                     "distinct_shapes": len(shapes),
                     "top_shapes": ", ".join(shapes.index[:4]) if len(shapes) <= 10 else "(free text)"})
type_audit = pd.DataFrame(rows)
display(type_audit[type_audit["top_shapes"] != "(free text)"].reset_index(drop=True))
```

```python
# --- results.time mixes two different quantities -------------------------------------
# The winner's row holds the total race time; everyone else holds a gap ("+5.478")
# or a lap count. One column, two meanings -> cannot become one number.
t = res_raw["time"][~is_missing(res_raw["time"])]
print("Gap to winner ('+...'):", int(t.str.startswith("+").sum()))
print("Total race time       :", int((~t.str.startswith("+")).sum()))
display(t.sample(6, random_state=SEED).to_frame())
```

**Finding:** the shapes give a clear conversion plan:

| Semantic type | Shape(s) | Columns | Target type |
|---|---|---|---|
| Date | `9999-99-99` | `races.date`, `races.*_date`, `drivers.dob` | `datetime64` |
| Clock time (time of day) | `99:99:99` | `races.time`, `races.*_time`, `pit_stops.time` | `timedelta` (time since midnight) |
| Duration | `9:99.999`, `99.999`, `9:99`, `9:99:99.999` | `qualifying.q1/q2/q3`, `*.fastestLapTime`, `lap_times.time`, `pit_stops.duration` | `float` seconds, in a new `<column>_sec` column |
| Number stored as text | `9`, `99`, `999.999` | e.g. `results.position`, `milliseconds`, `fastestLap`, `rank`, `drivers.number` | numeric |

Durations come in **several formats** (`ss.sss`, `m:ss.sss`, `mm:ss.sss`, `h:mm:ss.sss`, and `m:ss` without decimals), so the
parser must handle all of them. `results.time` / `sprint_results.time` mix total time and gaps: they are **left as text**,
because they are post-race and their numeric version already exists in `milliseconds`.

**Decision:** → **2.3** (typing), with the duration parser validated against the dataset's own `milliseconds` columns.

## 1.6 Are primary keys unique and foreign keys valid?
The keys were defined in Section 0.7. These two functions are reused after cleaning (2.5 and 3.x).

```python
# --- Key-audit functions ----------------------------------------------------------------
def pk_audit(tables):
    # Primary key: no duplicates, no missing values.
    rows = []
    for t, k in PRIMARY_KEYS.items():
        df = tables[t]
        rows.append({"table": t, "primary_key": ", ".join(k),
                     "duplicates": int(df.duplicated(k).sum()),
                     "nulls": int(df[k].apply(is_missing).any(axis=1).sum())})
    return pd.DataFrame(rows)

def fk_audit(tables):
    # Foreign key: every (non-missing) value exists in the parent table.
    rows = []
    for child, ccol, parent, pcol in FOREIGN_KEYS:
        c = tables[child][ccol]
        c = c[~is_missing(c)]
        orphans = ~to_num(c).isin(to_num(tables[parent][pcol]))
        rows.append({"foreign_key": f"{child}.{ccol}", "references": f"{parent}.{pcol}",
                     "orphans": int(orphans.sum())})
    return pd.DataFrame(rows)
```

```python
# --- Foreign keys (raw) ------------------------------------------------------------------------
fk_raw = fk_audit(raw)
print("Total orphan values:", fk_raw["orphans"].sum())
fk_raw
```

**Decision:** the same audit is re-run after typing (**2.5**); any duplicate primary key or orphan found there is removed and logged.

## 1.9 Do the values make sense together?
Sections 1.1–1.8 checked that values exist and have the right format. These checks test whether values are
**consistent with each other** (a race has exactly 3 podium places, a driver's age on race day is plausible, etc.).
The same function is re-run after cleaning in 3.7.

```python
# --- Consistency checks ------------------------------------------------------------------
# Each check returns the number of violating rows. Works on raw (text) and clean (typed)
# tables, because every value is converted with to_num / to_datetime inside.
def consistency_checks(tables):
    f = lambda s: to_num(s).astype("float64")
    races = tables["races"].assign(date=lambda d: pd.to_datetime(d["date"], errors="coerce"))
    res = tables["results"].merge(races[["raceId", "date"]], on="raceId", how="left")
    dob = pd.to_datetime(res["driverId"].map(tables["drivers"].set_index("driverId")["dob"]), errors="coerce")
    age = (res["date"] - dob).dt.days / 365.25
    pos, order, grid = f(res["position"]), f(res["positionOrder"]), f(res["grid"])
    podiums = (order <= PODIUM_CUTOFF).groupby(res["raceId"]).sum()
    starters = res.groupby("raceId").size()
    qual = tables["qualifying"]
    res_keys = res.set_index(["raceId", "driverId"]).index
    rows = [
        ("races", "season year differs from year of race date",
         (races["year"] != races["date"].dt.year).sum()),
        ("races", "duplicate (year, round)",
         races.duplicated(["year", "round"]).sum()),
        ("results", "driver age on race day outside 17-60",
         ((age < 17) | (age > 60)).sum()),
        ("results", "negative points or laps",
         ((f(res["points"]) < 0) | (f(res["laps"]) < 0)).sum()),
        ("results", "classified position differs from positionOrder",
         (pos.notna() & (pos != order)).sum()),
        ("results", "rows sharing a positionOrder with another driver",
         res.assign(_o=order).duplicated(["raceId", "_o"], keep=False).sum()),
        ("results", "races without exactly 3 podium rows",
         (podiums[starters >= 3] != 3).sum()),
        ("results", "rows sharing a grid slot (> 0) with another driver",
         res.assign(_g=grid)[grid > 0].duplicated(["raceId", "_g"], keep=False).sum()),
        ("qualifying", "duplicate qualifying position within a race",
         qual.duplicated(["raceId", "position"]).sum()),
        ("qualifying", "qualifying rows without a results row",
         (~qual.set_index(["raceId", "driverId"]).index.isin(res_keys)).sum()),
    ]
    return pd.DataFrame([(t, c, int(v)) for t, c, v in rows], columns=["table", "check", "violations"])

consistency_raw = consistency_checks(raw)
consistency_raw
```

# 2. Data Cleaning & Auditing

Every step below is justified by a finding from Section 1. The order matters:

| Step | What | Why this position |
|---|---|---|
| 2.1 | Work on copies | Raw tables must stay immutable |
| 2.2 | `\N` → `NaN` | Everything after needs real missing values |
| 2.3 | Semantic typing (numbers, dates, clock times, durations) | Needs 2.2 first, otherwise `\N` breaks the parsing |
| 2.4 | Drop (almost) empty columns | Measured on the typed data |
| 2.5 | Primary/foreign key audit | IDs are numeric now, so comparisons are exact |
| 2.6 | Remove out-of-scope rows (non-starters, Indy 500) | Before de-duplication, because some duplicates are non-starter entries |
| 2.7 | One row per `(raceId, driverId)`, then relief drivers (2.7b) | Only real race starts are left to choose between |
| 2.8 | Final checks, cleaning log, handover | Proves the result and documents it |

**Team decisions:** out-of-scope rows are **removed during cleaning**; **all 14 tables** are cleaned.

## 2.3 Semantic typing
Four conversions, following the plan from 1.4. Each one checks that **no non-missing value was lost** during parsing.

### 2.3a Numbers stored as text → numeric
A text column is converted **only if every non-missing value parses as a number**, so real text columns
(names, codes, `positionText` letters) are never touched. Whole numbers become the nullable integer type `Int64`,
which can hold missing values without turning into floats.

```python
# --- Generic numeric conversion -------------------------------------------------------------------
def numeric_if_possible(s):
    conv = pd.to_numeric(s, errors="coerce")
    if conv[s.notna()].isna().any():
        return None                       # at least one real text value -> leave the column alone
    if (conv.dropna() % 1 == 0).all():
        return conv.astype("Int64")       # whole numbers -> nullable integer
    return conv.astype("float64")

converted = []
for t, df in clean.items():
    for c in df.columns:
        if pd.api.types.is_numeric_dtype(df[c]) or df[c].isna().all():
            continue
        conv = numeric_if_possible(df[c])
        if conv is not None:
            df[c] = conv
            converted.append(f"{t}.{c}")
            log("2.3a numeric", t, conv.notna().sum(), f"{c}: text -> {conv.dtype}")
print(f"{len(converted)} columns converted:\n" + ", ".join(converted))
```

### 2.3b Dates → `datetime64`

```python
# --- Date columns ------------------------------------------------------------------------------------
# Strict format: any value that does not match YYYY-MM-DD would fail the check below.
DATE_COLUMNS = {
    "races":   ["date", "fp1_date", "fp2_date", "fp3_date", "quali_date", "sprint_date"],
    "drivers": ["dob"],
}
for t, cols in DATE_COLUMNS.items():
    for c in cols:
        if c not in clean[t]:
            continue
        before = clean[t][c].notna().sum()
        clean[t][c] = pd.to_datetime(clean[t][c], format="%Y-%m-%d", errors="coerce")
        lost = before - clean[t][c].notna().sum()
        assert lost == 0, f"{t}.{c}: {lost} values failed to parse"
        log("2.3b date", t, before, f"{c}: text -> datetime64")
clean["races"][["date", "quali_date"]].dtypes
```

### 2.3c Clock times → `timedelta`
These are times of day (e.g. race start `06:00:00`, stored in UTC), not durations. They become a `timedelta`
(time since midnight), which can be added to the matching date column when a full timestamp is needed.

```python
# --- Clock-time columns ----------------------------------------------------------------------------
CLOCK_COLUMNS = {
    "races":     ["time", "fp1_time", "fp2_time", "fp3_time", "quali_time", "sprint_time"],
    "pit_stops": ["time"],
}
for t, cols in CLOCK_COLUMNS.items():
    for c in cols:
        if c not in clean[t]:
            continue
        before = clean[t][c].notna().sum()
        clean[t][c] = pd.to_timedelta(clean[t][c], errors="coerce")
        lost = before - clean[t][c].notna().sum()
        assert lost == 0, f"{t}.{c}: {lost} values failed to parse"
        log("2.3c clock time", t, before, f"{c}: text -> timedelta (time of day)")
clean["races"][["date", "time"]].head(3)
```

### 2.3d Durations → seconds
One parser handles every format found in 1.4 by reading the parts from left to right
(`h:mm:ss.sss`, `m:ss.sss`, `m:ss`, `ss.sss`). The result goes into a new column named `<column>_sec`
so the unit is explicit, and the original text column is dropped.

```python
# --- Duration parser ---------------------------------------------------------------------------------
def duration_to_seconds(value):
    # "1:27.452" -> 87.452 | "49.111" -> 49.111 | "1:38" -> 98.0 | "1:02:03.5" -> 3723.5
    if pd.isna(value):
        return np.nan
    seconds = 0.0
    try:
        for part in str(value).split(":"):
            seconds = seconds * 60 + float(part)
    except ValueError:
        return np.nan
    return seconds

for example in ["1:27.452", "49.111", "1:38", "1:02:03.5"]:
    print(f"{example:>10} -> {duration_to_seconds(example)}")
```

```python
# --- Apply the parser ----------------------------------------------------------------------------------
DURATION_COLUMNS = {
    "qualifying":     ["q1", "q2", "q3"],
    "results":        ["fastestLapTime"],
    "sprint_results": ["fastestLapTime"],
    "lap_times":      ["time"],
    "pit_stops":      ["duration"],
}
for t, cols in DURATION_COLUMNS.items():
    for c in cols:
        if c not in clean[t]:
            continue
        before = clean[t][c].notna().sum()
        new_col = f"{c}_sec"
        clean[t][new_col] = clean[t][c].map(duration_to_seconds).astype("float64")
        lost = before - clean[t][new_col].notna().sum()
        assert lost == 0, f"{t}.{c}: {lost} values failed to parse"
        clean[t] = clean[t].drop(columns=c)
        log("2.3d duration", t, before, f"{c}: text -> {new_col} (seconds)")
clean["qualifying"].head(3)
```

```python
# --- Validate the parser against the dataset's own millisecond columns ------------------------------
# lap_times and pit_stops store each time twice (text + milliseconds). If the parser is
# right, parsed seconds == milliseconds / 1000.
for t, col in [("lap_times", "time_sec"), ("pit_stops", "duration_sec")]:
    df = clean[t].dropna(subset=[col, "milliseconds"])
    diff = (df[col] - df["milliseconds"] / 1000).abs()
    print(f"{t:<10} rows compared: {len(df):>7,} | exact (< 1 ms): {(diff < 0.001).mean():.4%} "
          f"| rows off by >= 1 ms: {(diff >= 0.001).sum()} | max difference: {diff.max():.3f} s")
```

**Result:** parsed seconds match `milliseconds / 1000` for (almost) every row, which proves the parser handles all formats.
Any remaining small differences come from text values stored **without** decimals (shape `9:99`) in the source data, i.e. the text
was rounded to whole seconds while `milliseconds` kept the exact value. This is a precision limit of the source, not a parsing error.

**Not converted, on purpose:** `results.time` and `sprint_results.time` mix total race time with gaps to the winner (1.4).
They stay as text; their numeric equivalent is `milliseconds`. Both are post-race and never used as features.

## 2.4 Drop (almost) empty columns
**Rule:** drop a column if **≥ 99 %** of its values are missing.
The threshold is deliberately high: `races.sprint_date` / `sprint_time` are ~98 % missing but are *structurally* missing
(only sprint weekends have a sprint), so they carry real information and must be kept.

```python
# --- Find and drop near-empty columns -------------------------------------------------------------------
EMPTY_THRESHOLD = 99.0
profile = missing_profile(clean)
to_drop = profile[profile["missing_%"] >= EMPTY_THRESHOLD]
display(to_drop.round(2))

for _, r in to_drop.iterrows():
    clean[r["table"]] = clean[r["table"]].drop(columns=r["column"])
    log("2.4 drop column", r["table"], r["rows"], f"{r['column']}: {r['missing_%']:.2f}% missing")
```

## 2.5 Primary and foreign keys
Same audit functions as 1.6, now on typed data. Any problem found is fixed and logged.

## 2.6 Remove out-of-scope rows
Based on 1.7. Each row of `results` is a prediction "will this driver finish on the podium in this race?",
which only makes sense for a driver who **took the start**:

| Removed | Rule | Why |
|---|---|---|
| Failed to qualify | `positionText == "F"` | Never started: no grid slot, nothing to predict |
| Withdrawn before the start | `positionText == "W"` and `laps == 0` | Never started |
| Indianapolis 500 (1950–1960) | race name `"Indianapolis 500"` | Separate event with its own drivers; distorts home-race and form statistics |

**Kept:** `grid = 0` with `laps > 0` (pit-lane starts). The Indy 500 is also removed from `qualifying` and `sprint_results`
(if present) so the race-entry tables stay consistent. `races` keeps it as a calendar record, and the standings tables are
official cumulative tables, so they are not changed.

```python
# --- Flag the out-of-scope rows ------------------------------------------------------------------------------
res = clean["results"]
indy_race_ids = clean["races"].loc[clean["races"]["name"].eq("Indianapolis 500"), "raceId"]

is_dnq       = res["positionText"].eq("F")
is_withdrawn = res["positionText"].eq("W") & res["laps"].fillna(0).eq(0)
is_indy      = res["raceId"].isin(indy_race_ids)

scope_counts = pd.Series({
    "failed to qualify (F)":    int(is_dnq.sum()),
    "withdrawn, 0 laps (W)":    int(is_withdrawn.sum()),
    "Indianapolis 500":         int(is_indy.sum()),
    "total removed (any rule)": int((is_dnq | is_withdrawn | is_indy).sum()),
}, name="rows")
scope_counts.to_frame()
```

```python
# --- Remove them ----------------------------------------------------------------------------------------------
rows_before_scope = len(res)
clean["results"] = res[~(is_dnq | is_withdrawn | is_indy)].reset_index(drop=True)
log("2.6 scope", "results", is_dnq.sum(), "removed: failed to qualify (positionText F)")
log("2.6 scope", "results", (is_withdrawn & ~is_dnq).sum(), "removed: withdrawn with 0 laps (positionText W)")
log("2.6 scope", "results", (is_indy & ~is_dnq & ~is_withdrawn).sum(), "removed: Indianapolis 500")

for t in ["qualifying", "sprint_results"]:
    indy_rows = clean[t]["raceId"].isin(indy_race_ids)
    if indy_rows.any():
        clean[t] = clean[t][~indy_rows].reset_index(drop=True)
        log("2.6 scope", t, indy_rows.sum(), "removed: Indianapolis 500")

print(f"results: {rows_before_scope:,} -> {len(clean['results']):,} rows")
```

## 2.7 One row per `(raceId, driverId)`
After 2.6, every remaining duplicate is a **real shared drive**: the driver started in one car and later drove a
teammate's car. We need one rule that picks exactly one row per pair. Three candidate rules are tested below.

```python
# --- Remaining duplicate pairs ---------------------------------------------------------------------------------
res = clean["results"]
pair_keys = ["raceId", "driverId"]
pairs = res[res.duplicated(pair_keys, keep=False)].copy()
n_pairs = pairs.groupby(pair_keys).ngroups
print(f"Rows involved: {len(pairs)} | pairs to resolve: {n_pairs}")
```

```python
# --- Candidate 1: "keep the row with a non-zero grid" ----------------------------------------------------------
# Works only if exactly one row of a pair has grid > 0.
resolved_by_grid = pairs.groupby(pair_keys)["grid"].apply(lambda g: (g > 0).sum() == 1)
print(f"Pairs this rule can resolve: {int(resolved_by_grid.sum())} of {n_pairs}")
print(f"Pairs where every row has grid > 0: {int(pairs.groupby(pair_keys)['grid'].apply(lambda g: (g > 0).all()).sum())}")
```

**Candidate 1 fails:** in shared drives, *both* rows carry a grid value (each car's starting slot), so "non-zero grid"
cannot tell them apart. We need a different signal for "the car he started in".

```python
# --- Candidates 2 and 3: "best finishing position" vs "earliest entry" --------------------------------------------
best_finish   = pairs.sort_values(["positionOrder", "resultId"]).groupby(pair_keys).head(1)
earliest_entry = pairs.sort_values("resultId").groupby(pair_keys).head(1)

rule_compare = pd.DataFrame({
    "podium rows kept": [int((best_finish["positionOrder"] <= PODIUM_CUTOFF).sum()),
                         int((earliest_entry["positionOrder"] <= PODIUM_CUTOFF).sum())],
    "rows kept":        [len(best_finish), len(earliest_entry)],
}, index=["best positionOrder", "earliest entry (lowest resultId)"])
podium_best     = best_finish.set_index(pair_keys)["positionOrder"].sort_index() <= PODIUM_CUTOFF
podium_earliest = earliest_entry.set_index(pair_keys)["positionOrder"].sort_index() <= PODIUM_CUTOFF
changed = podium_best != podium_earliest
print(f"Pairs whose podium label depends on the rule: {int(changed.sum())} of {n_pairs}")
rule_compare
```

```python
# --- Plot: podium rows kept by each rule ----------------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(7, 3))
ax.barh(rule_compare.index, rule_compare["podium rows kept"], color=[PALETTE[1], PALETTE[0]], height=0.5)
for i, v in enumerate(rule_compare["podium rows kept"]):
    ax.text(v + 0.3, i, str(v), va="center")
ax.set_xlabel("Podium rows among the de-duplicated pairs")
ax.set_title("Choosing by outcome inflates podiums")
plt.tight_layout()
plt.show()
```

**Candidate 2 (best `positionOrder`) is rejected:** it chooses the row *by its result*, so the target decides which row survives.
That is **selection on the target**: it systematically keeps the better outcome and inflates podiums (plot above). It can also
pair a driver with the grid slot of a car he did not start in, which corrupts the `grid` feature.

**Candidate 3 (earliest entry, lowest `resultId`)** does not look at the outcome at all. The check below shows *why*
the lowest `resultId` is the car the driver started in.

```python
# --- Evidence: the second row was appended later, out of sequence -------------------------------------------------
# For each race, take the resultId range of normal (non-duplicated) rows. The original entries
# sit inside that range; shared-drive rows added later have much higher ids.
normal_range = (res[~res.duplicated(pair_keys, keep=False)]
                .groupby("raceId")["resultId"].agg(["min", "max"]))
ev = pairs.join(normal_range, on="raceId")
ev["entry_order"] = ev.groupby(pair_keys)["resultId"].rank().astype(int).map({1: "1st entry", 2: "2nd entry", 3: "3rd entry"})
ev["inside_race_id_range"] = ev["resultId"].between(ev["min"], ev["max"])
pd.crosstab(ev["entry_order"], ev["inside_race_id_range"], margins=True)
```

```python
# --- A known example: 1957 British Grand Prix ---------------------------------------------------------------------
# Historically, Stirling Moss started from pole in car #18, retired, took over Tony Brooks's
# car #20 and won. Brooks started car #20 from 3rd. The earliest-entry rule should keep
# Moss -> car 18 / grid 1, and Brooks -> car 20 / grid 3.
race_1957 = clean["races"].query("year == 1957 and name == 'British Grand Prix'")["raceId"]
drivers_ = clean["drivers"].set_index("driverId")["driverRef"]
example = pairs[pairs["raceId"].isin(race_1957)].assign(driver=lambda d: d["driverId"].map(drivers_))
example = example.sort_values(["driver", "resultId"])
example["kept_by_earliest_entry"] = ~example.duplicated(pair_keys)
example[["driver", "resultId", "number", "grid", "positionOrder", "points", "laps", "kept_by_earliest_entry"]]
```

**Chosen rule: keep the earliest entry (lowest `resultId`) for each `(raceId, driverId)`.**

Why:
1. **It represents the car the driver started in**, which is exactly what a pre-race prediction is about: its `grid` is the
   driver's real starting position. Most first entries sit inside the race's normal id range while later rows were appended
   out of sequence (crosstab above), and the 1957 British GP shows the rule reproduces the known history.
2. **It never looks at the outcome**, so the label is not used to choose the row (unlike "best finish").
3. **It always resolves a pair** (ids are unique), unlike "non-zero grid".

**Trade-off:** the driver's result in the *second* car (e.g. Moss's 1957 win in Brooks's car) is not counted as his podium.
This affects only the pairs printed above (all before 1965), which is a small, documented effect.

```python
# --- Apply the rule and prove the grain ---------------------------------------------------------------------------
rows_before_dedup = len(res)
clean["results"] = (res.sort_values("resultId")
                       .drop_duplicates(pair_keys, keep="first")
                       .reset_index(drop=True))
log("2.7 grain", "results", rows_before_dedup - len(clean["results"]),
    "shared drives: kept earliest entry (lowest resultId) per (raceId, driverId)")

assert not clean["results"].duplicated(pair_keys).any()
print(f"results: {rows_before_dedup:,} -> {len(clean['results']):,} rows | (raceId, driverId) is unique ✔")
```

### 2.7b Relief drivers
In the 1950s a driver could take over a car mid-race **without having started the race in any car** (a "relief" driver).
His only row is the car another driver started, so after 2.7 that car's result appears twice: once for the driver who
started it and once for the relief driver (found in 1.9 as tied `positionOrder`).

**Rule:** within a race, if two drivers share a `positionOrder`, keep the earliest entry (the driver who started the car) and
remove the later one. **Why:** the relief driver never took the start, so there was no pre-race prediction to make for him, which is
the same reasoning as removing non-starters in 2.6. It also guarantees exactly three podium rows per race.

```python
# --- Find and remove relief-driver rows ------------------------------------------------------------
res = clean["results"]
tied = res[res.duplicated(["raceId", "positionOrder"], keep=False)].sort_values(["raceId", "positionOrder", "resultId"])
relief_ids = tied.loc[tied.duplicated(["raceId", "positionOrder"]), "resultId"]   # later entry of each tie
display(tied[["raceId", "driverId", "resultId", "number", "grid", "positionOrder", "points"]].head(8))

clean["results"] = res[~res["resultId"].isin(relief_ids)].reset_index(drop=True)
log("2.7 grain", "results", len(relief_ids),
    "relief drivers: removed later entry sharing a car's positionOrder with the driver who started it")

res = clean["results"]
podiums = (res["positionOrder"] <= PODIUM_CUTOFF).groupby(res["raceId"]).sum()
assert not res.duplicated(["raceId", "positionOrder"]).any()
assert (podiums[res.groupby("raceId").size() >= 3] == 3).all()
print(f"Removed {len(relief_ids)} relief-driver rows | every race now has exactly 3 podium rows ✔")
```

Each answer states the tables joined (multi-hop), the **grain** (what one row means), how **missingness** was handled,

## 4.1 Q1: Which circuits convert a front-row start into a podium most often, and does the ranking change after 2014?

**Approach**
- **Joins:** `results` → `races` (season, circuit) → `circuits` (circuit name), plus `qualifying` on `(raceId, driverId)` for the qualifying position.
- **Two definitions of "front row"**, as the question requires:
  - by **grid**: `grid` is 1 or 2 (where the car actually started, after penalties);
  - by **qualifying**: qualifying `position` is 1 or 2 (before penalties).
- **Conversion rate** = podiums ÷ front-row starts, per circuit and era (**pre-2014** vs **2014+**).
- **Sample:** only races that have qualifying data, so both definitions are measured on exactly the same races (1.3 showed qualifying
  starts in the mid-1990s). Circuits need at least `MIN_FRONT_ROW` front-row starts in an era to be ranked, to avoid rankings driven by 2–3 races.

```python
# --- Q1 base table: one row per driver-race, with circuit and qualifying position ----------------
q1 = (clean["results"][["raceId", "driverId", "grid", "positionOrder"]]
      .merge(clean["races"][["raceId", "year", "circuitId"]], on="raceId")
      .merge(clean["circuits"][["circuitId", "name"]].rename(columns={"name": "circuit"}), on="circuitId")
      .merge(clean["qualifying"][["raceId", "driverId", "position"]].rename(columns={"position": "quali_pos"}),
             on=["raceId", "driverId"], how="left"))

# Keep only races that have qualifying data, so both definitions use the same races.
q1 = q1[q1["raceId"].isin(clean["qualifying"]["raceId"])].copy()
q1["podium"] = q1["positionOrder"] <= PODIUM_CUTOFF
q1["era"] = np.where(q1["year"] >= HYBRID_ERA[0], "2014+", "pre-2014")
q1["front_row_grid"] = q1["grid"].isin([1, 2])
q1["front_row_quali"] = q1["quali_pos"].isin([1, 2])

assert not q1.duplicated(["raceId", "driverId"]).any()        # grain check
print(f"Rows: {len(q1):,} | races: {q1['raceId'].nunique()} | seasons {q1['year'].min()}-{q1['year'].max()}")
print(q1.groupby("era")["raceId"].nunique().rename("races").to_string())
```

```python
# --- Grid vs qualifying: do the two definitions disagree? --------------------------------------------
# Rows where they disagree are drivers moved by penalties (or pit-lane starts) or promoted
# because someone ahead was penalised.
groups = {
    "front row in both":                         q1["front_row_grid"] & q1["front_row_quali"],
    "qualified front row, started further back": ~q1["front_row_grid"] & q1["front_row_quali"],
    "started front row, qualified lower":        q1["front_row_grid"] & ~q1["front_row_quali"],
}
grid_vs_quali = pd.DataFrame({
    "rows": {k: int(m.sum()) for k, m in groups.items()},
    "podium_rate_%": {k: 100 * q1.loc[m, "podium"].mean() for k, m in groups.items()},
}).round(1)
grid_vs_quali
```

```python
# --- Conversion rate per circuit and era, for both definitions -------------------------------------------
MIN_FRONT_ROW = 10   # minimum front-row starts per circuit per era (= at least 5 races)

def conversion_by_circuit(df, flag):
    g = (df[df[flag]].groupby(["circuit", "era"])["podium"]
         .agg(starts="size", podiums="sum").unstack("era"))
    rate = 100 * g["podiums"] / g["starts"]
    out = pd.DataFrame({
        "starts_pre": g[("starts", "pre-2014")], "rate_pre_%": rate["pre-2014"],
        "starts_2014+": g[("starts", "2014+")], "rate_2014+_%": rate["2014+"],
    })
    ok = (out["starts_pre"] >= MIN_FRONT_ROW) & (out["starts_2014+"] >= MIN_FRONT_ROW)
    out = out[ok].copy()
    out["rank_pre"] = out["rate_pre_%"].rank(ascending=False, method="min").astype(int)
    out["rank_2014+"] = out["rate_2014+_%"].rank(ascending=False, method="min").astype(int)
    return out.sort_values("rate_2014+_%", ascending=False)

conv_grid = conversion_by_circuit(q1, "front_row_grid")
conv_quali = conversion_by_circuit(q1, "front_row_quali")
print(f"Circuits with >= {MIN_FRONT_ROW} front-row starts in both eras: {len(conv_grid)}")
conv_grid.round(1)
```

```python
# --- Does the ranking change? Rank correlation between eras --------------------------------------------------
# Spearman correlation compares the two rankings: +1 = identical order, 0 = unrelated order.
from scipy.stats import spearmanr

rows = []
for label, conv, flag in [("grid", conv_grid, "front_row_grid"), ("qualifying", conv_quali, "front_row_quali")]:
    rho, p = spearmanr(conv["rate_pre_%"], conv["rate_2014+_%"])
    overall = q1[q1[flag]].groupby("era")["podium"].mean() * 100
    rows.append({"definition": label, "circuits": len(conv),
                 "overall_pre_%": overall["pre-2014"], "overall_2014+_%": overall["2014+"],
                 "spearman_rho": rho, "p_value": p,
                 "top3_pre": ", ".join(conv.sort_values("rate_pre_%", ascending=False).index[:3]),
                 "top3_2014+": ", ".join(conv.index[:3])})
q1_summary = pd.DataFrame(rows).round(3)
q1_summary.T
```

## 4.2 Q2: Do drivers podium more often at home, after controlling for starting position?

**Approach**
- **Joins:** `results` → `drivers` (nationality) and `results` → `races` → `circuits` (country): a home race is one where the circuit's country matches the
  driver's nationality.
- **Mapping:** nationality ("British") and country ("UK") use different words, so an explicit mapping table is needed. Dual nationalities
  ("American-Italian") count as home in both countries; text is stripped of stray spaces and "United States"/"USA" are unified.
- **Controlling for starting position:** drivers are compared **within grid bands** (1, 2–3, 4–6, 7–10, 11+). Pit-lane starts (`grid = 0`) have no band and are excluded.
- **Test:** the **Cochran–Mantel–Haenszel (CMH)** method combines the home-vs-away comparison of every band into one **odds ratio**
  (> 1 = home advantage, < 1 = disadvantage, 1 = no effect), with a 95 % confidence interval and a p-value. It is written out below with
  `numpy`/`scipy` so every step is visible.

```python
# --- Nationality -> country mapping ---------------------------------------------------------------------
# Each nationality maps to the circuit country (or countries) that count as "home".
NATIONALITY_TO_COUNTRIES = {
    "American": ["USA"], "American-Italian": ["USA", "Italy"], "Argentine": ["Argentina"],
    "Argentinian": ["Argentina"], "Argentine-Italian": ["Argentina", "Italy"], "Australian": ["Australia"],
    "Austrian": ["Austria"], "Belgian": ["Belgium"], "Brazilian": ["Brazil"], "British": ["UK"],
    "Canadian": ["Canada"], "Chilean": ["Chile"], "Chinese": ["China"], "Colombian": ["Colombia"],
    "Czech": ["Czech Republic"], "Danish": ["Denmark"], "Dutch": ["Netherlands"], "East German": ["Germany"],
    "Finnish": ["Finland"], "French": ["France"], "German": ["Germany"], "Hungarian": ["Hungary"],
    "Indian": ["India"], "Indonesian": ["Indonesia"], "Irish": ["Ireland"], "Italian": ["Italy"],
    "Japanese": ["Japan"], "Liechtensteiner": ["Liechtenstein"], "Malaysian": ["Malaysia"], "Mexican": ["Mexico"],
    "Monegasque": ["Monaco"], "New Zealander": ["New Zealand"], "Polish": ["Poland"], "Portuguese": ["Portugal"],
    "Rhodesian": ["Rhodesia"], "Russian": ["Russia"], "South African": ["South Africa"], "Spanish": ["Spain"],
    "Swedish": ["Sweden"], "Swiss": ["Switzerland"], "Thai": ["Thailand"], "Uruguayan": ["Uruguay"],
    "Venezuelan": ["Venezuela"],
}
CIRCUIT_COUNTRY_ALIASES = {"United States": "USA"}

nationalities = set(clean["drivers"]["nationality"].str.strip())
unmapped = sorted(nationalities - set(NATIONALITY_TO_COUNTRIES))
print("Unmapped nationalities (treated as never home):", unmapped or "none ✔")

circuit_countries = set(clean["circuits"]["country"].str.strip().replace(CIRCUIT_COUNTRY_ALIASES))
mapped_countries = {c for cs in NATIONALITY_TO_COUNTRIES.values() for c in cs}
print("Circuit countries with no driver nationality (always 'away'):", sorted(circuit_countries - mapped_countries))
```

```python
# --- Q2 base table: one row per driver-race with a home flag and a grid band -------------------------------
q2 = (clean["results"][["raceId", "driverId", "grid", "positionOrder"]]
      .merge(clean["races"][["raceId", "year", "circuitId"]], on="raceId")
      .merge(clean["circuits"][["circuitId", "country"]], on="circuitId")
      .merge(clean["drivers"][["driverId", "nationality"]], on="driverId"))
q2["circuit_country"] = q2["country"].str.strip().replace(CIRCUIT_COUNTRY_ALIASES)
q2["home"] = [cc in NATIONALITY_TO_COUNTRIES.get(nat.strip(), [])
              for nat, cc in zip(q2["nationality"], q2["circuit_country"])]
q2["podium"] = q2["positionOrder"] <= PODIUM_CUTOFF

pit_lane = int((q2["grid"] == 0).sum())
q2 = q2[q2["grid"] > 0].copy()
q2["grid_band"] = pd.cut(q2["grid"], [0, 1, 3, 6, 10, np.inf], labels=["1", "2-3", "4-6", "7-10", "11+"])

assert not q2.duplicated(["raceId", "driverId"]).any()          # grain check
print(f"Rows: {len(q2):,} (excluded {pit_lane} pit-lane starts) | home rows: {int(q2['home'].sum()):,} ({q2['home'].mean():.1%})")
```

```python
# --- Why controlling for grid is needed: home and away drivers start from different places ------------------
start_mix = pd.crosstab(q2["grid_band"], q2["home"].map({True: "home", False: "away"}), normalize="columns") * 100
print("Mean grid: home", round(q2.loc[q2["home"], "grid"].mean(), 1), "| away", round(q2.loc[~q2["home"], "grid"].mean(), 1))
start_mix.round(1).rename_axis("grid band (% of starts)")
```

```python
# --- Podium rate by grid band, home vs away -------------------------------------------------------------------
by_band = (q2.groupby(["grid_band", "home"], observed=True)["podium"]
           .agg(starts="size", rate="mean").unstack("home"))
by_band.columns = [f"{m}_{'home' if h else 'away'}" for m, h in by_band.columns]
by_band[["rate_away", "rate_home"]] *= 100
by_band = by_band[["starts_away", "rate_away", "starts_home", "rate_home"]]
by_band.round(1)
```

```python
# --- Plot: podium rate by grid band, home vs away -------------------------------------------------------------
x = np.arange(len(by_band))
w = 0.38
fig, ax = plt.subplots(figsize=(10, 4.2))
ax.bar(x - w / 2, by_band["rate_away"], w, color=PALETTE[1], label="away")
ax.bar(x + w / 2, by_band["rate_home"], w, color=PALETTE[0], label="home")
for i, n in enumerate(by_band["starts_home"]):
    ax.text(i + w / 2, by_band["rate_home"].iloc[i] + 1, f"n={int(n)}", ha="center", fontsize=8)
ax.set_xticks(x, by_band.index)
ax.set_xlabel("Starting grid band")
ax.set_ylabel("Podium rate (%)")
ax.set_title("Q2: within the same grid band, home and away drivers podium at similar rates")
ax.legend(frameon=False)
plt.tight_layout()
plt.show()
```

```python
# --- Cochran-Mantel-Haenszel: one odds ratio across all strata ---------------------------------------------
# For each stratum (e.g. a grid band) build the 2x2 table:
#              podium   no podium
#   home         a          b
#   away         c          d
# The CMH odds ratio is a weighted average of the strata's odds ratios; the CI uses the
# Robins-Breslow-Greenland variance; the test is the CMH chi-square with continuity correction.
from scipy.stats import chi2

def cmh(df, strata, exposure="home", outcome="podium"):
    R = S = PR = PS_QR = QS = 0.0
    sum_a = sum_e = sum_v = 0.0
    for _, g in df.groupby(strata, observed=True):
        a = int((g[exposure] & g[outcome]).sum());  b = int((g[exposure] & ~g[outcome]).sum())
        c = int((~g[exposure] & g[outcome]).sum()); d = int((~g[exposure] & ~g[outcome]).sum())
        n = a + b + c + d
        if n < 2 or a + b == 0 or c + d == 0:
            continue                                   # stratum has no home or no away rows
        r, s = a * d / n, b * c / n
        p, q = (a + d) / n, (b + c) / n
        R += r; S += s; PR += p * r; PS_QR += p * s + q * r; QS += q * s
        sum_a += a
        sum_e += (a + b) * (a + c) / n
        sum_v += (a + b) * (c + d) * (a + c) * (b + d) / (n * n * (n - 1))
    odds_ratio = R / S
    se_log = np.sqrt(PR / (2 * R ** 2) + PS_QR / (2 * R * S) + QS / (2 * S ** 2))
    stat = (abs(sum_a - sum_e) - 0.5) ** 2 / sum_v
    return {"odds_ratio": odds_ratio,
            "ci_low": np.exp(np.log(odds_ratio) - 1.96 * se_log),
            "ci_high": np.exp(np.log(odds_ratio) + 1.96 * se_log),
            "p_value": chi2.sf(stat, df=1)}
```

```python
# --- Apply: no control, control for grid band, control for grid band and decade -----------------------------
q2["decade"] = q2["year"] // 10 * 10
q2["all"] = "all"
q2_results = pd.DataFrame({
    "no control (crude)":       cmh(q2, "all"),
    "controlling grid band":    cmh(q2, "grid_band"),
    "grid band + decade":       cmh(q2, ["grid_band", "decade"]),
}).T
q2_results.round(3)
```

### Answer Q2
*(Numbers below are from our run; they are printed by the cells above.)*

**No: there is no evidence of a home advantage once starting position is taken into account.**

- Without any control, home drivers even look slightly *worse* (crude odds ratio below 1). The reason is visible in the start-position table:
  home drivers start further back on average, because many home entries, especially in the 1950s–70s, were **local one-off drivers** in
  uncompetitive cars.
- Comparing drivers **within the same grid band** removes that effect: the CMH odds ratio moves close to 1 (≈ 0.9 in our run) and its 95 % confidence
  interval **includes 1**, with a p-value well above 0.05. Adding the decade as a second control (to account for era differences) gives the same conclusion.
- In the plot, the home and away bars are close in every band; the small differences go in both directions and have small home samples (see `n`).

**Grain:** one row per driver-race (starters only). **Missingness:** no missing nationalities or countries; pit-lane starts (`grid = 0`) are excluded because they
have no grid band; nationalities without an F1 circuit in their country simply never have a home race. The **Indianapolis 500** was already removed in 2.6, which
matters here: almost all its drivers were American racing in the USA and would have inflated "home" rows. **Cutoff:** both nationality and grid are known before the
race, so a home flag is a valid feature candidate, but this analysis suggests it adds little once grid is known.

## 4.3 Q3: Which constructors have the highest mechanical-retirement rate in 2014–2021 vs 2022–2024?

**Approach**
- **Joins:** `results` → `status` (why the race ended) → `constructors` (team name), and `results` → `races` (season → era).
- **Classifying the status values:** every status is put into one category. A **mechanical retirement** is a start that ended because a **car component
  failed** (engine, gearbox, brakes, hydraulics, power unit, etc.). Excluded, as the brief says: crashes and damage, driver causes (illness, injury, stalling),
  and regulatory decisions (disqualified, excluded, withdrawn). The full mapping is shown below.
- **Rate** = mechanical retirements ÷ **race starts** for that constructor in that era. After 2.6, every row in `results` is a start.
- Constructors need at least `MIN_STARTS` starts in an era (about one full season with two cars) to be ranked.

```python
# --- Status categories ------------------------------------------------------------------------------------------
# Finished = "Finished" or "+N Laps". Explicit lists for the non-mechanical categories;
# every other status names a failed car component and counts as mechanical.
ACCIDENT_DAMAGE = {"Accident", "Collision", "Collision damage", "Spun off", "Damage", "Debris",
                   "Puncture", "Tyre puncture", "Fatal accident"}
DRIVER = {"Physical", "Injured", "Injury", "Driver unwell", "Eye injury", "Illness", "Stalled"}
REGULATORY = {"Disqualified", "Excluded", "Withdrew", "Not classified", "107% Rule", "Did not qualify",
              "Did not prequalify", "Safety concerns", "Safety", "Not restarted", "Underweight"}
UNSPECIFIED = {"Retired", "Out of fuel", "Refuelling", "Fuel rig", "Handling"}

def status_category(status):
    s = str(status).strip()
    if s == "Finished" or s.startswith("+"):
        return "finished"
    if s in ACCIDENT_DAMAGE: return "accident / damage"
    if s in DRIVER:          return "driver"
    if s in REGULATORY:      return "regulatory"
    if s in UNSPECIFIED:     return "unspecified / other"
    return "mechanical"

status_map = clean["status"].assign(category=lambda d: d["status"].map(status_category))
status_map["category"].value_counts().to_frame("status values")
```

```python
# --- Q3 base table: starts in 2014-2024 with status category and team ---------------------------------------------
q3 = (clean["results"][["raceId", "driverId", "constructorId", "statusId"]]
      .merge(clean["races"][["raceId", "year"]], on="raceId")
      .merge(status_map[["statusId", "status", "category"]], on="statusId", how="left")
      .merge(clean["constructors"][["constructorId", "name"]].rename(columns={"name": "constructor"}), on="constructorId"))
q3 = q3[q3["year"].between(HYBRID_ERA[0], GROUND_EFFECT_ERA[1])].copy()
q3["era"] = np.where(q3["year"] <= HYBRID_ERA[1],
                     f"{HYBRID_ERA[0]}-{HYBRID_ERA[1]}", f"{GROUND_EFFECT_ERA[0]}-{GROUND_EFFECT_ERA[1]}")

assert q3["category"].notna().all()                              # every status was classified
assert not q3.duplicated(["raceId", "driverId"]).any()           # grain check
print(f"Starts: {len(q3):,}")
pd.crosstab(q3["category"], q3["era"], margins=True)
```

```python
# --- The mapping as it applies to these seasons (documented, as the brief asks) -------------------------------------
(q3.groupby(["category", "status"]).size().rename("starts").reset_index()
   .sort_values(["category", "starts"], ascending=[True, False])
   .set_index(["category", "status"]))
```

```python
# --- Mechanical-retirement rate per constructor and era --------------------------------------------------------------
MIN_STARTS = 40   # about one full season with two cars

q3_rates = (q3.groupby(["era", "constructor"])
              .agg(starts=("category", "size"),
                   mechanical=("category", lambda s: (s == "mechanical").sum()),
                   unspecified=("category", lambda s: (s == "unspecified / other").sum()))
              .reset_index())
q3_rates["mech_rate_%"] = 100 * q3_rates["mechanical"] / q3_rates["starts"]
# Sensitivity: upper bound if every unspecified retirement ("Retired", ...) was mechanical.
q3_rates["upper_bound_%"] = 100 * (q3_rates["mechanical"] + q3_rates["unspecified"]) / q3_rates["starts"]
q3_rates = q3_rates[q3_rates["starts"] >= MIN_STARTS].sort_values(["era", "mech_rate_%"], ascending=[True, False])
q3_rates.round(1).set_index(["era", "constructor"])
```

```python
# --- Plot: ranking per era (one panel per era, same scale) ------------------------------------------------------------
eras = sorted(q3_rates["era"].unique())
xmax = q3_rates["mech_rate_%"].max() * 1.25
fig, axes = plt.subplots(1, len(eras), figsize=(13, 0.32 * q3_rates.groupby("era").size().max() + 2))
for ax, era, color in zip(axes, eras, [PALETTE[1], PALETTE[0]]):
    d = q3_rates[q3_rates["era"] == era].sort_values("mech_rate_%")
    ax.barh(d["constructor"], d["mech_rate_%"], color=color, height=0.65)
    for i, (v, n) in enumerate(zip(d["mech_rate_%"], d["starts"])):
        ax.text(v + 0.2, i, f"{v:.1f}% (n={n})", va="center", fontsize=8)
    ax.set_xlim(0, xmax)
    ax.set_title(era)
    ax.set_xlabel("Mechanical retirements (% of starts)")
fig.suptitle("Q3: mechanical-retirement rate by constructor", fontweight="bold")
plt.tight_layout()
plt.show()
```

```python
# --- Overall reliability per era and the top 3 per era ------------------------------------------------------------------
overall = q3.groupby("era")["category"].apply(lambda s: 100 * (s == "mechanical").mean()).rename("mech_rate_%")
top3 = q3_rates.groupby("era").head(3).groupby("era")["constructor"].apply(", ".join).rename("top 3 (highest rate)")
pd.concat([overall.round(1), top3], axis=1)
```

*(Numbers below are from our run; they are printed by the cells above.)*

- **2014–2021 (hybrid era):** the highest mechanical-retirement rates belong to smaller or works teams struggling with the new power units,
  e.g. **Lotus F1**, **Toro Rosso**, **Renault** and **McLaren** (the McLaren-Honda years), at roughly 14–20 % of starts. **Mercedes**, the dominant team of the era,
  was among the most reliable (around 4 %).
- **2022–2024 (ground-effect era):** **Alfa Romeo** and **Alpine** have the highest rates (around 10–12 %), and the front-running teams (Red Bull, McLaren, Mercedes) are
  around 4 %.
- **Reliability improved overall:** the share of starts ending in a mechanical failure is clearly lower in 2022–2024 than in 2014–2021 (overall table above).

**Judgement calls (documented):** "Retired" without a stated cause, "Out of fuel" and similar are **not** counted as mechanical (the cause is unknown).
The `upper_bound_%` column shows the rate if they were: for most teams it changes little, but a team with several unexplained retirements
(e.g. RB in 2024) moves up noticeably, so its rank should be read with that caveat. Punctures are counted as damage (usually debris or contact),
while "Tyre", "Wheel" and "Wheel nut" failures count as mechanical (a component failed). Renamed teams (e.g. Toro Rosso → AlphaTauri → RB, Sauber → Alfa Romeo → Sauber)
are separate `constructorId`s in the data, so they appear separately.

**Grain:** one row per driver-race start; the rate is aggregated per constructor and era. **Missingness:** every status was classified (assertion above);
constructors with fewer than `MIN_STARTS` starts in an era are not ranked. **Cutoff:** `status` is post-race, so it is used here only for analysis; a team's
*past* reliability (e.g. mechanical DNF rate over previous races) is a valid, leakage-safe feature candidate.

# 5. Feature Selection & Engineering

_Leakage-safe grid, qualifying, standings (shifted), form (past races only) and context features, each justified with evidence; full leakage audit table._

**How features are chosen.** Every candidate has to pass three gates, in this order:

1. **Leakage gate (validity):** the value must be known *after qualifying and before lights out*. Only past races, qualifying of
   the current race and standings *before* the current race are allowed (Section 0.8 lists the forbidden columns).
2. **Relevance gate (evidence):** the feature must separate podium from non-podium rows — measured by single-feature ROC-AUC
   and by podium-rate curves, or by a conditional effect motivated by the data-engineering answers (Section 4).
3. **Redundancy gate:** if two features carry almost the same information (|Spearman ρ| > 0.95) the weaker one is dropped,
   unless an earlier finding shows that their *difference* matters.

**All evidence in this section is computed on seasons ≤ `TRAIN_END` (2019) only.** Validation and test seasons are never
looked at while choosing features, so the later scores stay an honest estimate.

## 5.1 Base table: one row per driver per race

```python
# --- Base table from the clean results ------------------------------------------------------------
MODEL_KEYS = ["raceId", "driverId"]

races_info = clean["races"][["raceId", "year", "round", "date", "circuitId"]]
base_df = (clean["results"][["resultId", "raceId", "driverId", "constructorId", "grid", "positionOrder", "position"]]
           .merge(races_info, on="raceId", how="left", validate="many_to_one"))
base_df[TARGET] = (base_df["positionOrder"] <= PODIUM_CUTOFF).astype(int)
base_df = base_df.sort_values(["date", "raceId", "grid"]).reset_index(drop=True)

assert not base_df.duplicated(MODEL_KEYS).any(), "grain broken"
assert base_df["date"].notna().all()
print(f"{len(base_df):,} driver-race rows | {base_df['raceId'].nunique():,} races | "
      f"{base_df['year'].min()}–{base_df['year'].max()} | podium rate {base_df[TARGET].mean():.1%}")
print("(raceId, driverId) is unique ✔")
```

`positionOrder` and `position` stay in this table **only** to build *past-race* form features (shifted, see 5.2c). They are
in `POST_RACE_COLUMNS`, so `assert_no_leakage()` will reject them if they ever end up in the feature list.

**Excluded on purpose (leakage gate):**

| Source | Why it is not a feature |
|---|---|
| `results.position`, `positionOrder`, `points`, `laps`, `time`, `milliseconds`, `statusId`, fastest-lap columns | outcome of the current race |
| `lap_times`, `pit_stops` (current race) | happen during the race |
| `driver_standings`, `constructor_standings` of the **current** race | include the current race's points → only the *previous* round is used (5.2e) |
| `constructor_results` of the current race | post-race points |

**Excluded on purpose (other reasons):**

| Candidate | Why it is not a feature |
|---|---|
| Home-race flag | Q2 (Section 4.2) found **no** home effect once grid is controlled for (CMH odds ratio ≈ 1) |
| `driverId`, `constructorId`, `circuitId` as categories | identities do not generalise in time: test-era teams/drivers barely exist in training (the *strength* of a team is captured by form and standings instead) |
| `year` | would let the model memorise eras instead of learning race-level signals |
| Sprint results | only exist from 2021, i.e. never in the training seasons |

## 5.2 Candidate features
Each function below adds one feature group. They only use columns that are known before the race, and every "history"
feature is computed with `shift(1)`, so the current race never contributes to its own feature value.

### 5.2a Grid group
The starting grid is published after qualifying (penalties included), so it is valid at the cutoff. A pit-lane start
(`grid = 0`, see 1.7) is put at the back of the field. `field_size` (number of starters) sets the base rate: with 3 podium
places, a smaller field means a higher chance for everyone.

```python
# --- Grid features ----------------------------------------------------------------------------------
def add_grid_features(df):
    df = df.copy()
    df["field_size"] = df.groupby("raceId")["driverId"].transform("size")
    df["pitlane_start"] = (df["grid"] == 0).astype(int)
    df["grid_pos"] = df["grid"].where(df["grid"] > 0, df["field_size"])     # pit lane = start last
    df["grid_pct"] = df.groupby("raceId")["grid_pos"].rank(method="min") / df["field_size"]
    return df
```

### 5.2b Qualifying group
* `quali_pos`: qualifying classification (pure pace, before penalties).
* `q1_gap_pct`: gap to the fastest Q1 time in %. **Q1 is used, not the best of Q1–Q3:** since 2006 every driver runs Q1,
  while Q2/Q3 times only exist for the faster drivers (1.2, handover), so a Q1 gap compares everyone on the same session.
  Before 2006 the single session is stored in `q1`, so the feature means the same thing in every format.
* `grid_penalty = grid_pos − quali_pos`: positions lost to penalties. Q1 (4.1) showed that penalised front-row qualifiers
  rarely convert, so this difference is kept explicitly.

```python
# --- Qualifying features ----------------------------------------------------------------------------
def add_qualifying_features(df, clean_tables):
    q = clean_tables["qualifying"][["raceId", "driverId", "position", "q1_sec"]].rename(columns={"position": "quali_pos"})
    q["q1_gap_pct"] = (q["q1_sec"] / q.groupby("raceId")["q1_sec"].transform("min") - 1) * 100
    df = df.merge(q[["raceId", "driverId", "quali_pos", "q1_gap_pct"]], on=MODEL_KEYS, how="left", validate="one_to_one")
    df["grid_penalty"] = df["grid_pos"] - df["quali_pos"]
    return df
```

### 5.2c Form group (leakage-safe: past races only)
Recent results of the driver and of the team over the last `w` races (window chosen empirically in 5.3). The current race is
excluded with `shift(1)` before the rolling window. Finishing position is divided by `field_size` so that "8th of 34" and
"8th of 16" are not treated the same across eras. A DNF is a race without a classified `position`.

```python
# --- Form features -----------------------------------------------------------------------------------
def past_mean(df, key, col, window):
    # Mean of `col` over the previous `window` rows of the same `key`; the current row is excluded by shift(1).
    return df.groupby(key)[col].transform(lambda s: s.shift(1).rolling(window, min_periods=1).mean())

def add_form_features(df, window):
    df = df.sort_values(["date", "raceId"]).copy()
    df["_finish_pct"] = df["positionOrder"] / df["field_size"]
    df["_dnf"] = df["position"].isna().astype(int)
    df["drv_podium_rate_last"] = past_mean(df, "driverId", TARGET, window)
    df["drv_avg_finish_pct_last"] = past_mean(df, "driverId", "_finish_pct", window)
    df["drv_dnf_rate_last"] = past_mean(df, "driverId", "_dnf", window)

    # Team level: aggregate both cars per race first, then roll over the team's previous races
    team = (df.groupby(["constructorId", "raceId"], sort=False)
              .agg(date=("date", "first"), podiums=(TARGET, "sum"), entries=(TARGET, "size"), fin=("_finish_pct", "mean"))
              .reset_index().sort_values(["date", "raceId"]))
    g = team.groupby("constructorId")
    past_sum = lambda col: g[col].transform(lambda s: s.shift(1).rolling(window, min_periods=1).sum())
    team["con_podium_rate_last"] = past_sum("podiums") / past_sum("entries")
    team["con_avg_finish_pct_last"] = g["fin"].transform(lambda s: s.shift(1).rolling(window, min_periods=1).mean())
    df = df.merge(team[["constructorId", "raceId", "con_podium_rate_last", "con_avg_finish_pct_last"]],
                  on=["constructorId", "raceId"], how="left", validate="many_to_one")
    return df.drop(columns=["_finish_pct", "_dnf"])
```

### 5.2d History group
Longer-term signals, also from past races only: number of previous starts (experience), career podium rate, and the
driver's podium rate at this circuit in earlier visits.

```python
# --- History features -----------------------------------------------------------------------------------
def add_history_features(df):
    df = df.sort_values(["date", "raceId"]).copy()
    by_driver = df.groupby("driverId")
    df["drv_starts_before"] = by_driver.cumcount()
    df["drv_career_podium_rate"] = by_driver[TARGET].transform(lambda s: s.shift(1).expanding().mean())
    df["drv_circuit_podium_rate"] = (df.groupby(["driverId", "circuitId"])[TARGET]
                                       .transform(lambda s: s.shift(1).expanding().mean()))
    return df
```

### 5.2e Standings group (shifted to *before* this race)
The standings tables are stored **after** each race (handover note). For a race in round *r* of a season, the feature is the
standing after the latest earlier round of the same season (`merge_asof` with `allow_exact_matches=False`, so round *r*
itself can never be used). In round 1 there are no standings yet: points share and wins are then truly 0, and the rank falls
back to the **final rank of the previous season**, which is also known before the season starts. Points are expressed as a
share of the leader's points, because the points system changed several times (e.g. 10 → 25 for a win in 2010).

```python
# --- Standings features ---------------------------------------------------------------------------------
def standings_before(df, standings, key, prefix, races_table):
    rr = races_table[["raceId", "year", "round"]]
    st = standings.merge(rr, on="raceId")[["year", "round", key, "points", "position", "wins"]]
    st["leader_pts"] = st.groupby(["year", "round"])["points"].transform("max")

    left = df[["raceId", "year", "round", key]].drop_duplicates().sort_values("round")
    m = pd.merge_asof(left, st.sort_values("round"), on="round", by=["year", key], allow_exact_matches=False)

    # Fallback for round 1 / first appearance in a season: final rank of the previous season
    final = st[st["round"].eq(st.groupby("year")["round"].transform("max"))][["year", key, "position"]]
    final = final.assign(year=final["year"] + 1).rename(columns={"position": "prev_season_rank"})
    m = m.merge(final.drop_duplicates(["year", key]), on=["year", key], how="left")

    m[f"{prefix}_rank_before"] = m["position"].fillna(m["prev_season_rank"])
    m[f"{prefix}_pts_share"] = (m["points"] / m["leader_pts"]).where(m["leader_pts"] > 0).fillna(0.0)
    m[f"{prefix}_wins_before"] = m["wins"].fillna(0)
    m["first_round"] = m["position"].isna().astype(int)
    cols = ["raceId", key, f"{prefix}_rank_before", f"{prefix}_pts_share", f"{prefix}_wins_before"]
    return m[cols + (["first_round"] if prefix == "drv" else [])]

def add_standings_features(df, clean_tables):
    df = df.merge(standings_before(df, clean_tables["driver_standings"], "driverId", "drv", clean_tables["races"]),
                  on=["raceId", "driverId"], how="left", validate="one_to_one")
    df = df.merge(standings_before(df, clean_tables["constructor_standings"], "constructorId", "con", clean_tables["races"]),
                  on=["raceId", "constructorId"], how="left", validate="many_to_one")
    return df
```

### 5.2f Circuit context
Q1 (4.1) showed that some circuits convert a front-row start into a podium much more often (hard to overtake). The feature is
that conversion rate **from earlier races at the same circuit only** (expanding mean, shifted), so no future race is used.

```python
# --- Circuit feature ------------------------------------------------------------------------------------
def add_circuit_features(df):
    fr = (df[df["grid_pos"] <= 2].groupby(["circuitId", "raceId"], sort=False)
            .agg(date=("date", "first"), conv=(TARGET, "mean")).reset_index().sort_values(["date", "raceId"]))
    fr["circuit_frontrow_conv_prior"] = fr.groupby("circuitId")["conv"].transform(lambda s: s.shift(1).expanding().mean())
    return df.merge(fr[["raceId", "circuit_frontrow_conv_prior"]], on="raceId", how="left", validate="many_to_one")
```

```python
# --- One function that builds every candidate feature ----------------------------------------------------
def build_features(base, clean_tables, form_window):
    df = add_grid_features(base)
    df = add_qualifying_features(df, clean_tables)
    df = add_form_features(df, form_window)
    df = add_history_features(df)
    df = add_standings_features(df, clean_tables)
    df = add_circuit_features(df)
    return df.sort_values(["date", "raceId", "grid_pos"]).reset_index(drop=True)

CANDIDATE_GROUPS = {
    "grid":       ["grid_pos", "grid_pct", "pitlane_start", "field_size"],
    "qualifying": ["quali_pos", "q1_gap_pct", "grid_penalty"],
    "form":       ["drv_podium_rate_last", "drv_avg_finish_pct_last", "drv_dnf_rate_last",
                   "con_podium_rate_last", "con_avg_finish_pct_last"],
    "history":    ["drv_starts_before", "drv_career_podium_rate", "drv_circuit_podium_rate"],
    "standings":  ["drv_rank_before", "drv_pts_share", "drv_wins_before", "first_round",
                   "con_rank_before", "con_pts_share", "con_wins_before"],
    "circuit":    ["circuit_frontrow_conv_prior"],
}
CANDIDATES = [c for cols in CANDIDATE_GROUPS.values() for c in cols]
assert_no_leakage(CANDIDATES)
print(f"{len(CANDIDATES)} candidate features in {len(CANDIDATE_GROUPS)} groups")
```

## 5.3 Choosing the form window (experiment)
How many past races should "recent form" cover? Too short is noisy, too long mixes different cars (a season has ~20 races).
The form features are rebuilt for several windows and scored by single-feature ROC-AUC on the training seasons.

```python
# --- Form window experiment (training seasons only) -------------------------------------------------------
from sklearn.metrics import roc_auc_score

def single_auc(df, col):
    # ROC-AUC of one feature on its own; direction-free (a feature where low = podium also counts).
    m = df[col].notna()
    if df.loc[m, col].nunique() < 2:
        return np.nan
    a = roc_auc_score(df.loc[m, TARGET], df.loc[m, col])
    return max(a, 1 - a)

FORM_COLS = CANDIDATE_GROUPS["form"]
window_rows = []
for w in [1, 3, 5, 10, 20]:
    tmp = add_form_features(add_grid_features(base_df), w)
    tmp = tmp[tmp["year"] <= TRAIN_END]
    window_rows.append({"window": w, **{c: single_auc(tmp, c) for c in FORM_COLS}})
window_df = pd.DataFrame(window_rows).set_index("window")
window_df.round(3)
```

```python
# --- Plot: AUC by window ------------------------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(9, 4))
for c in FORM_COLS:
    ax.plot(window_df.index, window_df[c], marker="o", label=c.replace("_last", ""))
ax.axvline(10, color="grey", ls="--", lw=1)
ax.text(10.4, ax.get_ylim()[0] + 0.01, "chosen: 10", color="grey")
ax.set_xscale("log"); ax.set_xticks(window_df.index); ax.set_xticklabels(window_df.index)
ax.set_xlabel("Window (previous races)"); ax.set_ylabel("Single-feature ROC-AUC (train seasons)")
ax.set_title("Form features: longer windows help, with diminishing returns")
ax.legend(fontsize=8, loc="lower right")
plt.tight_layout(); plt.show()
```

```python
FORM_WINDOW = 10
gain = (window_df.loc[10] - window_df.loc[5]).mean(), (window_df.loc[20] - window_df.loc[10]).mean()
print(f"Mean AUC gain 5→10: {gain[0]:+.3f} | 10→20: {gain[1]:+.3f}  → FORM_WINDOW = {FORM_WINDOW}")
```

**Finding:** every form feature improves from 1 to 10 races, and then the curve flattens. **10 races** is chosen: it is the
elbow, and it covers about half a season, while 20 races would regularly reach into the previous season's car.

## 5.4 Build the candidate table and test it for leakage

```python
# --- Build all candidate features ---------------------------------------------------------------------------
cand_df = build_features(base_df, clean, FORM_WINDOW)
assert len(cand_df) == len(base_df) and not cand_df.duplicated(MODEL_KEYS).any()
print(f"Candidate table: {cand_df.shape[0]:,} rows × {len(CANDIDATES)} candidate features | grain unique ✔")
cand_df[MODEL_KEYS + ["year", "round"] + CANDIDATES].head()
```

```python
# --- Leakage test 1: the future cannot change the past ------------------------------------------------------
# Rebuild every feature using ONLY data before a cut-off date. If any feature looked into the future, the values of races
# before the cut-off would change. They must be identical.
CUT = pd.Timestamp("2010-01-01")
early_ids = set(races_info.loc[races_info["date"] < CUT, "raceId"])
clean_early = {k: (v[v["raceId"].isin(early_ids)] if "raceId" in v.columns else v) for k, v in clean.items()}

feat_early = build_features(base_df[base_df["raceId"].isin(early_ids)], clean_early, FORM_WINDOW)
a = cand_df[cand_df["raceId"].isin(early_ids)].set_index(MODEL_KEYS).sort_index()[CANDIDATES]
b = feat_early.set_index(MODEL_KEYS).sort_index()[CANDIDATES]
pd.testing.assert_frame_equal(a, b, check_dtype=False)
print(f"Features of {len(a):,} pre-2010 rows are identical when all later data is removed ✔")
```

```python
# --- Leakage test 2: history features of a driver's first start are empty -----------------------------------
first_start = cand_df.sort_values("date").groupby("driverId").head(1)
assert first_start[["drv_podium_rate_last", "drv_career_podium_rate", "drv_circuit_podium_rate"]].isna().all().all()
assert (first_start["drv_starts_before"] == 0).all()

# --- Leakage test 3: standings come from the previous round, not the current one ----------------------------
rr = clean["races"][["raceId", "year", "round"]]
prev_round = rr.assign(round=rr["round"] + 1).rename(columns={"raceId": "prev_raceId"})
st_prev = clean["driver_standings"][["raceId", "driverId", "position"]].rename(
    columns={"raceId": "prev_raceId", "position": "rank_after_prev_round"})
chk = (cand_df.merge(prev_round, on=["year", "round"], how="inner")
              .merge(st_prev, on=["prev_raceId", "driverId"], how="inner"))
assert (chk["drv_rank_before"] == chk["rank_after_prev_round"]).all()
print(f"drv_rank_before equals the official rank after the previous round in all {len(chk):,} checkable rows ✔")
print("No history at a driver's first start ✔")
```

```python
# --- Leakage test 4: the current race's result cannot change its own features --------------------------------
# Test 1 shows that no feature uses LATER races, but it cannot catch a feature that uses the CURRENT race (that race is in
# both builds). Here the finishing order of one race is reversed and every feature is rebuilt: the features of that race's
# own rows must not change. Repeated for 5 random races from the training seasons.
rng_leak = np.random.default_rng(SEED)
feat_ref = cand_df.set_index(MODEL_KEYS)
changed = {}
for race in rng_leak.choice(base_df.loc[base_df["year"].between(2003, TRAIN_END), "raceId"].unique(), 5, replace=False):
    pert = base_df.copy()
    m = pert["raceId"] == race
    pert.loc[m, "positionOrder"] = pert.loc[m, "positionOrder"].to_numpy()[::-1]          # reverse the result
    pert.loc[m, "position"] = pert.loc[m, "positionOrder"].where(rng_leak.random(m.sum()) < 0.5)  # random DNFs
    pert[TARGET] = (pert["positionOrder"] <= PODIUM_CUTOFF).astype(int)
    feat_pert = build_features(pert, clean, FORM_WINDOW).set_index(MODEL_KEYS)
    keys = pert.loc[m].set_index(MODEL_KEYS).index
    diff = (feat_ref.loc[keys, CANDIDATES].fillna(-999) != feat_pert.loc[keys, CANDIDATES].fillna(-999)).sum()
    for col, n in diff[diff > 0].items():
        changed[col] = changed.get(col, 0) + int(n)
assert not changed, f"features use the current race's result: {changed}"
print("Reversing a race's result changes none of that race's own features (5 races) ✔")
```

**Finding:** all four tests pass. Test 1 is the strongest: rebuilding every feature *without* any data after 2010 gives
exactly the same values for the races before 2010, so no feature uses information from later races. Test 4 covers the other direction: the result of the race being predicted never feeds into its own features.

## 5.6 Relevance: does each feature separate podium from non-podium?

```python
# --- Single-feature ROC-AUC on training seasons ------------------------------------------------------------------
evid = cand_df[cand_df["year"] <= TRAIN_END]                        # evidence data: never val/test
evid_q = evid[evid["year"] >= 2003]                                 # full qualifying coverage (5.5)
group_of = {c: g for g, cols in CANDIDATE_GROUPS.items() for c in cols}

relevance = pd.DataFrame({
    "group": [group_of[c] for c in CANDIDATES],
    "auc_1950_2019": [single_auc(evid, c) for c in CANDIDATES],
    "auc_2003_2019": [single_auc(evid_q, c) for c in CANDIDATES],
    "direction": ["higher → podium" if evid[[c, TARGET]].corr("spearman").iloc[0, 1] > 0 else "lower → podium"
                  for c in CANDIDATES],
}, index=CANDIDATES).sort_values("auc_2003_2019", ascending=False)
relevance.round(3)
```

```python
# --- Plot: single-feature AUC, coloured by group ---------------------------------------------------------------
group_colour = dict(zip(CANDIDATE_GROUPS, PALETTE))
fig, ax = plt.subplots(figsize=(9, 7))
ax.barh(relevance.index, relevance["auc_2003_2019"], color=[group_colour[g] for g in relevance["group"]], height=0.7)
ax.scatter(relevance["auc_1950_2019"], relevance.index, color="black", s=12, zorder=3, label="AUC 1950–2019")
ax.axvline(0.55, color="grey", ls="--", lw=1)
ax.text(0.553, len(relevance) - 0.6, "relevance threshold 0.55", color="grey", fontsize=8)
ax.invert_yaxis(); ax.set_xlim(0.45, 0.95)
ax.set_xlabel("Single-feature ROC-AUC (bars: 2003–2019, dots: 1950–2019)")
ax.set_title("How well each candidate separates podium rows on its own")
handles = [plt.Rectangle((0, 0), 1, 1, color=group_colour[g]) for g in CANDIDATE_GROUPS]
ax.legend(handles + [ax.collections[0]], list(CANDIDATE_GROUPS) + ["AUC 1950–2019"], fontsize=8, loc="lower right")
plt.tight_layout(); plt.show()
```

```python
# --- Plot: podium rate across the range of each continuous feature (training seasons ≥ 2003) ----------------------
cont = [c for c in CANDIDATES if evid_q[c].nunique() > 10]
ncol = 5
nrow = int(np.ceil(len(cont) / ncol))
fig, axes = plt.subplots(nrow, ncol, figsize=(15, 2.6 * nrow), sharey=True)
for ax, c in zip(axes.flat, cont):
    d = evid_q[[c, TARGET]].dropna()
    bins = pd.qcut(d[c], 10, duplicates="drop")
    rate = d.groupby(bins, observed=True)[TARGET].mean() * 100
    mids = d.groupby(bins, observed=True)[c].median()
    ax.plot(mids, rate, marker="o", ms=3, color=group_colour[group_of[c]])
    ax.set_title(c, fontsize=9); ax.tick_params(labelsize=7)
for ax in axes.flat[len(cont):]:
    ax.axis("off")
for ax in axes[:, 0]:
    ax.set_ylabel("Podium rate (%)")
fig.suptitle("Podium rate by decile of each feature (training seasons 2003–2019)", fontweight="bold")
plt.tight_layout(); plt.show()
```

**Finding:**
* **Qualifying and grid are the strongest single signals** (AUC ≈ 0.91 in 2003–2019): podium rate falls steeply with
  starting position. In the full-qualifying era the two are almost equal, because they differ only where a penalty was applied.
* **Form, standings and history are strong too** (AUC ≈ 0.75–0.88): they describe how competitive the driver and car are.
* **Weak on their own:** `pitlane_start`, `first_round`, `circuit_frontrow_conv_prior`, `grid_penalty`, `field_size`
  (AUC ≈ 0.50–0.56). Before dropping them, 5.7 checks whether they matter *conditionally*.
* The ranking is stable between all seasons (dots) and the full-qualifying era (bars).

## 5.7 Conditional evidence for the weak features
A feature can be weak overall but important for a subgroup. Three of the weak candidates were motivated by Section 4, so they
are checked where they should matter.

```python
# --- Conditional checks ------------------------------------------------------------------------------------------------
top3_q = evid_q[evid_q["quali_pos"] <= 3]
pen = (top3_q.assign(penalised=np.where(top3_q["grid_penalty"] > 0, "penalised", "no penalty"))
             .groupby("penalised")[TARGET].agg(podium_rate="mean", rows="size"))

def circuit_quartiles(d):
    d = d[d["grid_pos"] <= 2].dropna(subset=["circuit_frontrow_conv_prior"])
    q = pd.qcut(d["circuit_frontrow_conv_prior"], 4, labels=["Q1", "Q2", "Q3", "Q4"])
    return d.groupby(q, observed=True)[TARGET].agg(podium_rate="mean", rows="size")
circ = pd.concat({"1950–2002": circuit_quartiles(evid[evid["year"] < 2003]), "2003–2019": circuit_quartiles(evid_q)}, axis=1)

fs_bins = pd.cut(evid["field_size"], [0, 18, 20, 22, 24, 26, 40])
fs = evid.groupby(fs_bins, observed=True).agg(podium_rate=(TARGET, "mean"), rows=(TARGET, "size"),
                                              mean_field=("field_size", "mean"))

fig, axes = plt.subplots(1, 3, figsize=(15, 3.8))
axes[0].bar(pen.index, pen["podium_rate"] * 100, color=[PALETTE[0] if i == "no penalty" else PALETTE[1] for i in pen.index])
axes[0].set_ylim(0, 80)
for i, (r, n) in enumerate(zip(pen["podium_rate"], pen["rows"])):
    axes[0].text(i, r * 100 + 1, f"{r:.0%} (n={n})", ha="center")
axes[0].set_title("Top-3 qualifiers: grid penalty"); axes[0].set_ylabel("Podium rate (%)")
x = np.arange(len(circ))
for k, (era, col) in enumerate([("1950–2002", PALETTE[6]), ("2003–2019", PALETTE[5])]):
    axes[1].bar(x + (k - 0.5) * 0.38, circ[(era, "podium_rate")] * 100, width=0.38, color=col, label=era)
axes[1].set_xticks(x); axes[1].set_xticklabels(circ.index)
axes[1].set_title("Front-row starters: prior circuit conversion"); axes[1].set_xlabel("quartile: low → high prior conversion")
axes[1].legend(fontsize=8)
axes[2].bar([str(i) for i in fs.index], fs["podium_rate"] * 100, color=PALETTE[3])
axes[2].plot([str(i) for i in fs.index], 300 / fs["mean_field"], color="black", ls="--", lw=1, marker="o", ms=3,
             label="3 / mean field size")
axes[2].set_title("Field size"); axes[2].tick_params(axis="x", rotation=30); axes[2].legend(fontsize=8)
plt.tight_layout(); plt.show()
display(pen.round(3), circ.round(3), fs.round(3))
```

**Finding:**
* **`grid_penalty` matters conditionally:** among top-3 qualifiers, a penalty cuts the podium rate sharply (left), as Q1 found.
  It is weak overall only because most drivers have no penalty → **kept**.
* **`circuit_frontrow_conv_prior`:** before 2003 front-row starters converted somewhat more often at circuits with a high
  past conversion, but in **2003–2019 the effect is gone** (flat bars, middle). Q1 describes *which* circuits converted in the

# 6. Train / Validation / Test Split

_Temporal split: train ≤ 2019, validation 2020–2021, test ≥ 2022 — with justification._

## 6.1 Why a chronological split
The split boundaries were fixed in the setup (0.3): **train ≤ 2019, validation 2020–2021, test 2022–2024**. They are kept,
and the reasons for a split by season (rather than a random split of rows) are:

1. **It mirrors real use.** The model is used to predict *future* races from the past. A test on later seasons is the honest
   version of that question; a random split would test on races that happened *before* some training races.
2. **Whole races stay together.** Exactly 3 drivers per race finish on the podium, so rows of the same race depend on each
   other. A random split would put drivers of the same race in train and test.
3. **No look-ahead through form features.** Form and standings summarise earlier races; with a random split, the training
   set would contain races that come *after* test races of the same driver.
4. **It tests the hard case: a regulation change.** The test seasons (2022–2024) are the new ground-effect era (Q3, 4.3),
   validation (2020–2021) is the end of the hybrid era. The test score therefore shows how well the model holds up when the
   cars change, which is what a real user would face.

What was **not** fixed in the setup is where training *starts*. That is decided below with evidence.

## 6.2 Where should training start? (experiment)
Hypothesis from 5.5: before 2003 the qualifying group is fully (pre-1994) or largely (1996–2002) missing, and older eras have
different field sizes and podium rates (1.8). Older seasons add rows but may teach patterns that no longer hold.

**Experiment:** train the same two simple models on different start years (always up to 2019) and score them on the
**validation** seasons only. The test seasons are not touched.
* Logistic regression (median imputation + missing indicators + scaling): the linear baseline.
* Histogram gradient boosting (handles missing values natively): a non-linear check, so the decision does not depend on one model type.

Validation has only about 120 podium rows, so small differences can be noise. A **race-level bootstrap** (resampling whole
validation races 300 times) gives a 95 % interval for each difference against the 2003 start.

```python
# --- Training-window experiment (validation seasons only) -------------------------------------------------------------
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import HistGradientBoostingClassifier
from sklearn.metrics import average_precision_score
warnings.filterwarnings("ignore", message="Unknown solver options")   # harmless scipy/sklearn version notice

def quick_models():
    return {
        "logreg": make_pipeline(SimpleImputer(strategy="median", add_indicator=True), StandardScaler(),
                                LogisticRegression(max_iter=2000)),
        # early_stopping=False: sklearn switches early stopping on automatically above 10,000 rows, which would make
        # large and small training windows train differently (a confound we found while running this experiment)
        "hist_gb": HistGradientBoostingClassifier(learning_rate=0.05, max_iter=300, early_stopping=False, random_state=SEED),
    }

val_mask = model_df["year"].between(*VAL_YEARS)
X_val_raw, y_val_raw = model_df.loc[val_mask, FEATURES], model_df.loc[val_mask, TARGET]

window_results, window_preds = [], {}
for start in [1950, 1970, 1980, 1994, 2003, 2010]:
    tr = model_df[model_df["year"].between(start, TRAIN_END)]
    for name, model in quick_models().items():
        model.fit(tr[FEATURES], tr[TARGET])
        p = model.predict_proba(X_val_raw)[:, 1]
        window_preds[(start, name)] = p
        window_results.append({"train_start": start, "model": name, "train_rows": len(tr),
                               "quali_missing_%": 100 * tr["quali_pos"].isna().mean(),
                               "val_roc_auc": roc_auc_score(y_val_raw, p), "val_pr_auc": average_precision_score(y_val_raw, p)})
window_results = pd.DataFrame(window_results)
window_results.pivot(index=["train_start", "train_rows", "quali_missing_%"], columns="model",
                     values=["val_roc_auc", "val_pr_auc"]).round(3)
```

```python
# --- Race-level bootstrap: is any start year really better than 2003? -------------------------------------------------
rng = np.random.default_rng(SEED)
val_race_ids = model_df.loc[val_mask, "raceId"].to_numpy()
y_val_arr = y_val_raw.to_numpy()
race_rows = {r: np.flatnonzero(val_race_ids == r) for r in np.unique(val_race_ids)}
boot_idx = [np.concatenate([race_rows[r] for r in rng.choice(list(race_rows), len(race_rows))]) for _ in range(300)]

boot_rows = []
for (start, name), p in window_preds.items():
    if start == 2003:
        continue
    ref = window_preds[(2003, name)]
    d = [average_precision_score(y_val_arr[b], p[b]) - average_precision_score(y_val_arr[b], ref[b]) for b in boot_idx]
    boot_rows.append({"train_start": start, "model": name, "pr_auc_diff_vs_2003": np.mean(d),
                      "ci_low": np.percentile(d, 2.5), "ci_high": np.percentile(d, 97.5)})
boot_df = pd.DataFrame(boot_rows)
boot_df["significant"] = (boot_df["ci_low"] > 0) | (boot_df["ci_high"] < 0)
boot_df.round(3)
```

```python
# --- Plot: validation scores by training start year ------------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(12, 3.8), sharex=True)
for ax, metric in zip(axes, ["val_pr_auc", "val_roc_auc"]):
    for name, col in [("logreg", PALETTE[0]), ("hist_gb", PALETTE[1])]:
        d = window_results[window_results["model"] == name]
        ax.plot(d["train_start"], d[metric], marker="o", color=col, label=name)
    ax.axvline(2003, color="grey", ls="--", lw=1)
    ax.set_title(metric.replace("val_", "Validation ").replace("_", "-").upper())
    ax.set_xlabel("First training season (last = 2019)")
axes[0].legend()
fig.suptitle("Does adding older seasons help? (scored on 2020–2021)", fontweight="bold")
plt.tight_layout(); plt.show()
```

**Finding:** the hypothesis that old seasons *hurt* is **not** confirmed. Validation scores are nearly flat across start years
(logistic regression ≈ 0.93 ROC-AUC / ≈ 0.73 PR-AUC for every window), and **no earlier start year is significantly better
or worse than 2003** in the bootstrap (every interval includes 0). Adding 50 years of data neither helps nor hurts. Only
starting *later* (2010) is significantly worse for gradient boosting: with ~4,300 rows it has too little data.

> A first version of this experiment showed gradient boosting doing clearly better with all seasons. That came from sklearn
> switching on early stopping only for the large windows; with early stopping off for every window the gap disappears.

**Decision: `TRAIN_START = 2003`.** Since validation cannot separate the windows, the decision rests on data quality:
1. **Qualifying features are real, not imputed.** From 2003 qualifying coverage is complete (≈ 0.3 % missing vs 18–63 % for
   earlier starts). Qualifying is the strongest group (5.6); with an earlier start most of its training values would be
   imputed, which also makes the ablation (Section 9) and the explanations (Section 10) harder to interpret.
2. **Closer to the test era.** The 2003+ seasons share the modern formats (single-lap and later knockout qualifying, points
   down to 8th or 10th place), so training rows look more like the rows the model will be used on.
3. **No cost on validation**, as shown above, while ~6,800 training rows are still plenty for 19 features (2010 shows where
   too few rows start to hurt).

## 6.3 The final split

```python
# --- Split by season -----------------------------------------------------------------------------------------------------
def assign_split(year):
    if TRAIN_START <= year <= TRAIN_END:
        return "train"
    if VAL_YEARS[0] <= year <= VAL_YEARS[1]:
        return "val"
    if year >= TEST_START:
        return "test"
    return "unused"            # seasons before TRAIN_START

model_df["split"] = model_df["year"].map(assign_split)
train_df = model_df[model_df["split"] == "train"].reset_index(drop=True)
val_df = model_df[model_df["split"] == "val"].reset_index(drop=True)
test_df = model_df[model_df["split"] == "test"].reset_index(drop=True)

# Checks: no season or race in two sets, and the sets are in time order
assert set(train_df["raceId"]).isdisjoint(val_df["raceId"]) and set(val_df["raceId"]).isdisjoint(test_df["raceId"])
assert train_df["date"].max() < val_df["date"].min() and val_df["date"].max() < test_df["date"].min()

split_summary = (model_df[model_df["split"] != "unused"].groupby("split", sort=False)
                 .agg(seasons=("year", lambda y: f"{y.min()}–{y.max()}"), races=("raceId", "nunique"),
                      rows=("raceId", "size"), podium_rate=(TARGET, "mean"))
                 .loc[["train", "val", "test"]])
split_summary["share_of_rows"] = split_summary["rows"] / split_summary["rows"].sum()
print("No race appears in two sets ✔ | train < val < test in time ✔")
split_summary.round(3)
```

**Finding:** about 76 % / 9 % / 15 % of rows (train / val / test). The podium rate is similar in all three sets (14–15 %), so

## 6.4 What a random split would hide (experiment)
To show *why* the random split is rejected, the same logistic regression is scored two ways: on a random 20 % of the
2003–2021 rows, and on the real chronological validation seasons.

```python
# --- Random vs chronological validation ------------------------------------------------------------------------------------
from sklearn.model_selection import train_test_split

pool = model_df[model_df["year"].between(TRAIN_START, VAL_YEARS[1])]
tr_r, va_r = train_test_split(pool, test_size=0.2, random_state=SEED, stratify=pool[TARGET])
split_cmp = []
for label, tr, va in [("random 80/20 (2003–2021)", tr_r, va_r), ("chronological (val 2020–2021)", train_df, val_df)]:
    for name, model in quick_models().items():
        model.fit(tr[FEATURES], tr[TARGET])
        p = model.predict_proba(va[FEATURES])[:, 1]
        split_cmp.append({"split": label, "model": name, "roc_auc": roc_auc_score(va[TARGET], p),
                          "pr_auc": average_precision_score(va[TARGET], p),
                          "races_in_both": len(set(tr["raceId"]) & set(va["raceId"]))})
pd.DataFrame(split_cmp).round(3)
```

**Finding:** the two scores are close (the random split is a little higher for gradient boosting, a little lower for logistic

## 7.2 The pre-processing pipeline

```python
# --- Building blocks --------------------------------------------------------------------------------------------------
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer
from sklearn.impute import MissingIndicator
from sklearn.metrics import f1_score, precision_recall_curve

class QuantileCapper(BaseEstimator, TransformerMixin):
    """Caps each column at a quantile learned in fit() (NaN stays NaN, so imputation still sees it)."""
    def __init__(self, upper=0.99):
        self.upper = upper
    def fit(self, X, y=None):
        self.cap_ = np.nanquantile(np.asarray(X, dtype=float), self.upper, axis=0)
        return self
    def transform(self, X):
        return np.minimum(np.asarray(X, dtype=float), self.cap_)
    def get_feature_names_out(self, input_features=None):
        return np.asarray(input_features, dtype=object)

CAP_COLS = ["q1_gap_pct"]
LOG_COLS = ["drv_starts_before", "drv_wins_before", "con_wins_before"]

def make_preprocessor(features, cap_cols=CAP_COLS, log_cols=LOG_COLS, add_indicators=True):
    """Pre-processing for any subset of FEATURES (the ablation in Section 9 reuses it)."""
    cap = [c for c in features if c in cap_cols]
    log = [c for c in features if c in log_cols]
    rest = [c for c in features if c not in cap + log]
    impute_scale = [("impute", SimpleImputer(strategy="median")), ("scale", StandardScaler())]
    branches = [
        ("cap", Pipeline([("cap", QuantileCapper(0.99))] + impute_scale), cap),
        ("log", Pipeline([("log", FunctionTransformer(np.log1p, feature_names_out="one-to-one"))] + impute_scale), log),
        ("num", Pipeline(impute_scale), rest),
    ]
    if add_indicators:
        # error_on_new=False: a column that was never missing in training (e.g. drv_pts_share) may still be missing
        # in a new record (inference demo, Section 11); it is then imputed instead of raising an error.
        branches.append(("miss", MissingIndicator(features="missing-only", error_on_new=False), list(features)))
    return ColumnTransformer([b for b in branches if b[2]], verbose_feature_names_out=False)

def fit_transform_splits(features, preprocessor=None, fit_df=None):
    """Fit on the training rows (default) and transform train/val/test. Returns DataFrames with feature names."""
    pre = preprocessor if preprocessor is not None else make_preprocessor(features)
    pre.fit((fit_df if fit_df is not None else train_df)[features])
    names = [n.replace("missingindicator_", "missing_") for n in pre.get_feature_names_out()]
    out = [pd.DataFrame(pre.transform(d[features]).astype(float), columns=names, index=d.index)
           for d in (train_df, val_df, test_df)]
    return pre, *out
```

## 7.4 Helper for the experiments
All experiments use the same simple, fixed model (logistic regression) and are scored on the **validation** seasons with the
metrics required by the brief: ROC-AUC, PR-AUC and F1. The F1 threshold is the one that maximises F1 on the **training**
predictions, so the validation and test seasons are never used to pick it.

**For Section 8:** this helper picks the threshold on *training* predictions, which is fine for logistic regression here. For random forests, gradient boosting and the neural network, training probabilities are over-confident, so the final threshold must be chosen on the **validation** predictions (still never on test).

```python
# --- Evaluation helper --------------------------------------------------------------------------------------------------------
def best_f1_threshold(y_true, p):
    prec, rec, thr = precision_recall_curve(y_true, p)
    f1 = 2 * prec * rec / np.clip(prec + rec, 1e-12, None)
    return thr[np.nanargmax(f1[:-1])]

def evaluate(Xtr, ytr, Xva, yva, model=None):
    model = model if model is not None else LogisticRegression(max_iter=2000)
    model.fit(Xtr, ytr)
    p_tr, p_va = model.predict_proba(Xtr)[:, 1], model.predict_proba(Xva)[:, 1]
    thr = best_f1_threshold(ytr, p_tr)
    return {"val_roc_auc": roc_auc_score(yva, p_va), "val_pr_auc": average_precision_score(yva, p_va),
            "val_f1": f1_score(yva, (p_va >= thr).astype(int)), "threshold": thr}

baseline_scores = evaluate(X_train, y_train, X_val, y_val)
pd.Series(baseline_scores, name="our pipeline (logistic regression)").round(3).to_frame()
```

## 7.5 Preprocessing-order experiments
### E1: split first, then fit the pre-processing — or the other way round?
**Ours:** split by season first, fit imputation/cap/scaling on training rows only.
**Alternative:** fit the same pre-processing on *all* seasons (train + val + test) before splitting — a common mistake.

```python
# --- E1: fit scope -----------------------------------------------------------------------------------------------------------
all_rows = pd.concat([train_df, val_df, test_df])
pre_leaky, Xtr_l, Xva_l, Xte_l = fit_transform_splits(FEATURES, fit_df=all_rows)

def learned_values(pre):
    num_cols = [c for c in FEATURES if c not in CAP_COLS + LOG_COLS]
    b = pre.named_transformers_
    num = lambda step, attr, col: getattr(b["num"].named_steps[step], attr)[num_cols.index(col)]
    return {"q1_gap_pct cap": b["cap"].named_steps["cap"].cap_[0],
            "q1_gap_pct scaling mean": b["cap"].named_steps["scale"].mean_[0],
            "grid_pos scaling mean": num("scale", "mean_", "grid_pos"),
            "drv_rank_before imputation median": num("impute", "statistics_", "drv_rank_before"),
            "con_pts_share scaling mean": num("scale", "mean_", "con_pts_share")}

e1 = pd.DataFrame({
    "split → fit (ours)": {**learned_values(PREPROCESSOR), **baseline_scores},
    "fit on all → split (leaky)": {**learned_values(pre_leaky), **evaluate(Xtr_l, y_train, Xva_l, y_val)},
})
e1.round(4)
```

**Result E1:** fitting on all seasons changes the learned values (the cap, scaling means and imputation medians now contain
2020–2024 information) while the validation scores barely move. The small score difference is **not** a reason to accept it:
the leaky version's validation and test scores are no longer an honest estimate, because the transformation already "knew"
the distribution of the seasons being predicted. In real use those seasons do not exist yet. → **split first, fit on train
only** (our pipeline).

### E2: cap before scaling — or scale before capping?
**Ours:** cap `q1_gap_pct` at the training p99 **before** computing the scaling mean/std.
**Alternative:** impute and scale first, then cap the scaled values (same quantile). Then the outliers have already inflated
the standard deviation used for scaling.

```python
# --- E2: order of capping and scaling ---------------------------------------------------------------------------------------------
def make_preprocessor_scale_then_cap(features):
    pre = make_preprocessor(features)
    cap = [c for c in features if c in CAP_COLS]
    pre.transformers[0] = ("cap", Pipeline([("impute", SimpleImputer(strategy="median")), ("scale", StandardScaler()),
                                            ("cap", QuantileCapper(0.99))]), cap)
    return pre

pre_e2, Xtr_e2, Xva_e2, Xte_e2 = fit_transform_splits(FEATURES, preprocessor=make_preprocessor_scale_then_cap(FEATURES))
e2 = pd.DataFrame({
    "cap → scale (ours)": {"scaling std of raw q1_gap_pct": PREPROCESSOR.named_transformers_["cap"].named_steps["scale"].scale_[0],
                           "q1_gap_pct range after (train)": X_train["q1_gap_pct"].max() - X_train["q1_gap_pct"].min(),
                           "q1_gap_pct std after (train)": X_train["q1_gap_pct"].std(), **baseline_scores},
    "scale → cap": {"scaling std of raw q1_gap_pct": pre_e2.named_transformers_["cap"].named_steps["scale"].scale_[0],
                    "q1_gap_pct range after (train)": Xtr_e2["q1_gap_pct"].max() - Xtr_e2["q1_gap_pct"].min(),
                    "q1_gap_pct std after (train)": Xtr_e2["q1_gap_pct"].std(), **evaluate(Xtr_e2, y_train, Xva_e2, y_val)},
})
e2.round(4)
```

```python
# --- Plot: q1_gap_pct after both orders ------------------------------------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 3.5))
bins = np.linspace(min(X_train["q1_gap_pct"].min(), Xtr_e2["q1_gap_pct"].min()),
                   max(X_train["q1_gap_pct"].max(), Xtr_e2["q1_gap_pct"].max()), 60)
ax.hist(X_train["q1_gap_pct"], bins=bins, alpha=0.6, color=PALETTE[0], label="cap → scale (ours)")
ax.hist(Xtr_e2["q1_gap_pct"], bins=bins, alpha=0.6, color=PALETTE[1], label="scale → cap")
ax.set_xlabel("q1_gap_pct after pre-processing"); ax.set_ylabel("Training rows"); ax.legend()
ax.set_title("E2: scaling before capping squeezes the feature into a narrow band")
plt.tight_layout(); plt.show()
```

**Result E2:** when scaling comes first, the few extreme Q1 gaps inflate the scaling standard deviation (1.81 instead of 1.51),
so the feature ends up squeezed (std ≈ 0.84 instead of 1, plot) and capping afterwards cannot undo it. ROC-AUC and PR-AUC
are unchanged, because both orders keep the *ranking* of the rows; F1 is slightly lower. The bigger effect is on anything that
reads the scale of a feature: regularisation strength, neural-network inputs, and the coefficients and explanations in
Sections 8–10, where a squeezed feature looks less important than it is. → **cap before scaling** (our pipeline).

### E3: how to handle missing values
Not an order question, but the choice of missing-value strategy is checked the same way:
(a) median + missing indicators (ours), (b) median only, (c) drop training rows with any missing value
(validation rows cannot be dropped at prediction time, so they are still imputed).

```python
# --- E3: missing-value strategy ---------------------------------------------------------------------------------------------------
_, Xtr_b, Xva_b, _ = fit_transform_splits(FEATURES, preprocessor=make_preprocessor(FEATURES, add_indicators=False))
complete = train_df[FEATURES].notna().all(axis=1).to_numpy()
pre_c = make_preprocessor(FEATURES, add_indicators=False).fit(train_df.loc[complete, FEATURES])
e3 = pd.DataFrame({
    "(a) median + indicators (ours)": {**baseline_scores, "train_rows": len(X_train)},
    "(b) median only": {**evaluate(Xtr_b, y_train, Xva_b, y_val), "train_rows": len(Xtr_b)},
    "(c) drop incomplete train rows": {**evaluate(pre_c.transform(train_df.loc[complete, FEATURES]), y_train[complete],
                                                  pre_c.transform(val_df[FEATURES]), y_val), "train_rows": int(complete.sum())},
})
e3.round(4)
```

**Result E3:** the three strategies score within ≈ 0.006 of each other (well inside the validation noise measured in 6.2):
(a) is best on ROC-AUC and F1, (b) is slightly best on PR-AUC. The decision therefore rests on the data:
* **(c) is rejected:** it throws away 26 % of the training rows — mostly rookies and new teams, exactly the drivers whose history
  is missing — so the model never learns how to treat them, although they appear in every new season (and must be imputed at

* **(a) over (b):** the indicators keep the fact that a value was imputed ("no history yet"), which the tree models and the

# 8. Predictive Modeling

**Goal:** predict `podium` with at least two statistical models and a shallow neural network, compare them fairly, and
report ROC-AUC, PR-AUC, F1, precision and recall on train, validation and test.

**Rules for this section** (from the brief and the 7.7 handover):
- Every model uses the **same rows** (`X_train` / `X_val` / `X_test`, train 2003–2019, val 2020–21, test 2022–24) and the
  **same 31 columns** (19 features + 12 missing indicators), so their scores are directly comparable.
- **Hyperparameters and the decision threshold are chosen on validation only.** The test set is scored **once**, in 8.7,
  after every choice is frozen.
- **Primary metric: PR-AUC.** Podiums are only ≈14 % of rows; PR-AUC measures how well the podium rows are ranked at the top
  and its "coin-flip" value is the podium rate (≈0.15), not 0.5. **Secondary: ROC-AUC** (ranking quality over all rows,
  easy to compare with the single-feature AUCs of 5.6). F1, precision and recall describe the yes/no decisions at the
  chosen threshold.

## 8.1 Helpers

### 8.1a Imports
The model classes and metric functions used in this section. Some were already imported in Sections 6–7; importing them
again here keeps Section 8 readable on its own.

```python
# --- Imports for Section 8 ----------------------------------------------------------------------------------------------
from sklearn.metrics import (roc_auc_score, average_precision_score, f1_score, precision_score, recall_score,
                             confusion_matrix, roc_curve, precision_recall_curve)
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, HistGradientBoostingClassifier
```

### 8.1b Metric functions
- `scores_at` computes all five metrics. ROC-AUC and PR-AUC only use the *order* of the probabilities (no threshold);
  F1, precision and recall need a yes/no decision, so they take a threshold `thr`.
- `val_scores` is a short version used while tuning: only the two ranking metrics, on validation.

```python
# --- Metric functions ----------------------------------------------------------------------------------------------------
def scores_at(y, p, thr):
    """The five metrics for true labels y and probabilities p, with yes/no decisions taken at threshold thr."""
    y_hat = (p >= thr).astype(int)                         # 1 = predicted podium
    return {"roc_auc": roc_auc_score(y, p),                  # ranking quality over all rows
            "pr_auc": average_precision_score(y, p),         # ranking quality of the rare class (primary metric)
            "f1": f1_score(y, y_hat),                        # balance of precision and recall
            "precision": precision_score(y, y_hat, zero_division=0),   # of predicted podiums, share that were right
            "recall": recall_score(y, y_hat)}                # of real podiums, share that were found

def val_scores(p_val):
    """Ranking metrics on the validation seasons (used to compare settings while tuning)."""
    return {"val_pr_auc": average_precision_score(y_val, p_val), "val_roc_auc": roc_auc_score(y_val, p_val)}
```

### 8.1c A record of every final model
Three dictionaries collect the final version of each model so later cells (comparison, plots, Sections 9–10) can loop
over them. `register` stores a model, its probabilities on train and validation, and its **threshold chosen on
validation** with `best_f1_threshold` (the helper from 7.4, here applied to validation as the handover requires).
Test probabilities are deliberately *not* stored yet; they are added once, in 8.7.

```python
# --- Shared records of the final models -----------------------------------------------------------------------------------
models = {}          # name -> fitted model (contract name from the plan)
PROBS = {}           # name -> {"train": probabilities, "val": probabilities}; "test" is added once, in 8.7
THRESHOLDS = {}      # name -> decision threshold, chosen on validation (the one that maximises F1)

def register(name, model, p_train, p_val):
    """Store a final model, its train/val probabilities and its validation-chosen threshold."""
    models[name] = model
    PROBS[name] = {"train": p_train, "val": p_val}
    THRESHOLDS[name] = best_f1_threshold(y_val, p_val)     # threshold picked on VALIDATION, never on test
    print(f"{name:<22} val PR-AUC {average_precision_score(y_val, p_val):.3f} | "
          f"val ROC-AUC {roc_auc_score(y_val, p_val):.3f} | threshold (val, max F1) {THRESHOLDS[name]:.3f}")

# Reminder of the class balance the metrics are working with
print(f"Train: {len(y_train)} rows, {int(y_train.sum())} podiums ({y_train.mean():.1%}) | "
      f"validation: {len(y_val)} rows, {int(y_val.sum())} podiums ({y_val.mean():.1%})")
```

## 8.2 Rule baseline: "the top 3 on the grid finish on the podium"

Before any learning, the simplest possible predictor: rank drivers by starting position. Its "probability" is
`1 / grid_pos`, which is larger for better starts; only the order matters for ROC-AUC and PR-AUC, and the threshold chosen
on validation turns it into a yes/no rule. Every model below must beat this, otherwise it has learned nothing beyond the grid.
The rule reads the raw `grid_pos` from `train_df` / `val_df` (not the scaled matrix), so it is written as a tiny class.

```python
# --- Rule baseline ---------------------------------------------------------------------------------------------------------
class GridRule:
    """Not a learned model: scores each driver by 1 / starting position (pit-lane starts already count as last)."""
    def predict_from_df(self, df):
        return 1.0 / df["grid_pos"].to_numpy(dtype=float)

grid_rule = GridRule()
register("grid rule (baseline)", grid_rule, grid_rule.predict_from_df(train_df), grid_rule.predict_from_df(val_df))

# Translate the threshold back into grid positions: 1 / threshold = the last grid slot predicted to podium
print(f"→ the validation threshold means: predict podium if grid_pos ≤ {1 / THRESHOLDS['grid rule (baseline)']:.0f}")
```

## 8.3 Logistic regression (statistical baseline)

**How it works:** it multiplies every (scaled) column by a weight, adds them up, and squeezes the sum through a sigmoid
into a probability between 0 and 1. Training finds the weights that best separate podium from non-podium rows.

### 8.3a Tuning on validation
Two settings are tried, 8 combinations in total:
- `C`: how strongly the weights are kept small (regularisation). Smaller `C` = simpler model.
- `class_weight`: `None` treats every row equally; `"balanced"` makes a missed podium count ≈6× more than a false alarm,
  to compensate for podiums being rare.

Each combination is trained on the training seasons and scored on validation.

```python
# --- Logistic regression: try 8 settings, score each on validation ----------------------------------------------------------
lr_rows = []
for C in [0.01, 0.1, 1, 10]:
    for cw in [None, "balanced"]:
        m = LogisticRegression(C=C, class_weight=cw, max_iter=5000).fit(X_train, y_train)
        lr_rows.append({"C": C, "class_weight": str(cw), **val_scores(m.predict_proba(X_val)[:, 1])})

# Best setting first (highest validation PR-AUC)
lr_grid = pd.DataFrame(lr_rows).sort_values("val_pr_auc", ascending=False).reset_index(drop=True)
lr_grid.round(4)
```

**Finding:** all eight settings land within ≈0.007 PR-AUC of each other, so logistic regression is not sensitive to its
settings here. Class re-weighting does not improve the ranking (PR-AUC); it only shifts the probabilities upward, which the
validation threshold compensates for anyway.

### 8.3b Final logistic regression
The best row of the table above is refitted and registered.

```python
# --- Final logistic regression: the best setting from 8.3a ---------------------------------------------------------------------
best = lr_grid.iloc[0]
logreg = LogisticRegression(C=best["C"],
                            class_weight=None if best["class_weight"] == "None" else "balanced",
                            max_iter=5000).fit(X_train, y_train)
register("logistic regression", logreg, logreg.predict_proba(X_train)[:, 1], logreg.predict_proba(X_val)[:, 1])
```

## 8.4 Random forest

**How it works:** 500 decision trees, each grown on a random bootstrap sample of the training rows and considering a random
subset of the columns at every split. The forest's probability is the share of trees that vote "podium". Averaging many
different trees cancels out the mistakes of individual trees.

### 8.4a Tuning on validation
Two settings that limit how much a single tree can memorise, 9 combinations:
- `max_depth`: the maximum number of questions a tree can ask (`None` = unlimited).
- `min_samples_leaf`: the minimum number of training rows in each final leaf (1 = a leaf may describe a single driver-race).

`class_weight="balanced_subsample"` re-weights podiums inside each tree's bootstrap sample. This cell takes ≈20 seconds.

```python
# --- Random forest: try 9 settings, score each on validation -------------------------------------------------------------------
rf_rows = []
for max_depth in [None, 8, 12]:
    for min_leaf in [1, 5, 20]:
        m = RandomForestClassifier(n_estimators=500, max_depth=max_depth, min_samples_leaf=min_leaf,
                                   class_weight="balanced_subsample", n_jobs=-1, random_state=SEED)
        m.fit(X_train, y_train)
        rf_rows.append({"max_depth": str(max_depth), "min_samples_leaf": min_leaf,
                        **val_scores(m.predict_proba(X_val)[:, 1])})

rf_grid = pd.DataFrame(rf_rows).sort_values("val_pr_auc", ascending=False).reset_index(drop=True)
rf_grid.round(4)
```

**Finding:** fully grown trees (`min_samples_leaf = 1`) score clearly worse on validation: they memorise individual
training races. Larger leaves (20 rows) generalise best, i.e. this data rewards **simple** trees.

### 8.4b Final random forest

```python
# --- Final random forest: the best setting from 8.4a ------------------------------------------------------------------------------
best = rf_grid.iloc[0]
rf = RandomForestClassifier(n_estimators=500,
                            max_depth=None if best["max_depth"] == "None" else int(best["max_depth"]),
                            min_samples_leaf=int(best["min_samples_leaf"]),
                            class_weight="balanced_subsample", n_jobs=-1, random_state=SEED).fit(X_train, y_train)
register("random forest", rf, rf.predict_proba(X_train)[:, 1], rf.predict_proba(X_val)[:, 1])
```

## 8.5 Gradient boosting

**How it works:** small trees are built one after another; each new tree is trained to correct the errors the previous
trees still make, and its contribution is shrunk by the learning rate. `HistGradientBoostingClassifier` is scikit-learn's
version of the same idea as LightGBM / XGBoost.

### 8.5a Tuning on validation
Three settings, 18 combinations:
- `learning_rate`: how big a step each new tree takes (smaller = slower, more careful learning).
- `max_leaf_nodes`: the size of each tree (4 leaves = very simple trees).
- `max_iter`: the number of trees.

Fixed: at least 20 rows per leaf, a small L2 penalty, and **no early stopping**, so the number of trees is an explicit,
tuned setting (6.2 found that sklearn's automatic early stopping distorted a comparison).

```python
# --- Gradient boosting: try 18 settings, score each on validation -----------------------------------------------------------------
gb_rows = []
for lr_ in [0.02, 0.05]:
    for leaves in [4, 7, 15]:
        for n_iter in [100, 200, 400]:
            m = HistGradientBoostingClassifier(learning_rate=lr_, max_leaf_nodes=leaves, max_iter=n_iter,
                                               min_samples_leaf=20, l2_regularization=1.0, early_stopping=False,
                                               random_state=SEED)
            m.fit(X_train, y_train)
            gb_rows.append({"learning_rate": lr_, "max_leaf_nodes": leaves, "max_iter": n_iter,
                            **val_scores(m.predict_proba(X_val)[:, 1])})

gb_grid = pd.DataFrame(gb_rows).sort_values("val_pr_auc", ascending=False).reset_index(drop=True)
print("For comparison: the default setting used in 6.2 / 7.6 (learning rate 0.05, 31 leaves, 300 trees) "
      "scored val PR-AUC ≈ 0.69")
gb_grid.head(8).round(4)                                   # the 8 best settings
```

**Finding:** boosting only catches up with logistic regression when it is kept **small and slow** (few leaves, low
learning rate). The default, larger setting overfits. Same message as the forest: most of the signal is simple and
close to monotone (better grid / form / standings → more podiums), which a linear model already captures.

### 8.5b Final gradient boosting

```python
# --- Final gradient boosting: the best setting from 8.5a ----------------------------------------------------------------------------
best = gb_grid.iloc[0]
gb = HistGradientBoostingClassifier(learning_rate=best["learning_rate"], max_leaf_nodes=int(best["max_leaf_nodes"]),
                                    max_iter=int(best["max_iter"]), min_samples_leaf=20, l2_regularization=1.0,
                                    early_stopping=False, random_state=SEED).fit(X_train, y_train)
register("gradient boosting", gb, gb.predict_proba(X_train)[:, 1], gb.predict_proba(X_val)[:, 1])
```

## 8.6 Shallow neural network (Keras)

**How it works:** a feed-forward network. The 31 input columns go into one or two hidden layers of units; each unit
computes a weighted sum of its inputs followed by ReLU (`max(0, x)`), which lets the network learn curved relationships.
A final unit with a sigmoid turns the result into the podium probability. The weights are learned by gradient descent
(Adam) on the binary cross-entropy loss.

**Over-fitting controls:**
- **Dropout (30 %):** during training, a random 30 % of the hidden units are switched off at every step, so the network
  cannot rely on any single unit.
- **L2 penalty:** large weights are penalised, which keeps the network smooth.
- **Early stopping:** after each pass over the data (epoch) the network is scored on validation PR-AUC; training stops when
  it has not improved for 20 epochs, and the weights of the best epoch are restored.

### 8.6a Setup: data as arrays, seeds
Keras works with numpy arrays of type float32. A neural network's result also depends on its random starting weights, so
every architecture is trained with **5 seeds**; only a difference larger than the seed-to-seed spread means something.

```python
# --- Keras setup ---------------------------------------------------------------------------------------------------------------------
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"                    # hide TensorFlow's start-up messages
import tensorflow as tf
from tensorflow import keras

NN_SEEDS = [SEED, 1, 2, 3, 4]                              # the project seed plus four more
X_train_np, X_val_np, X_test_np = (X.to_numpy("float32") for X in (X_train, X_val, X_test))
print(f"Keras {keras.__version__} | input columns: {X_train_np.shape[1]}")
```

### 8.6b Building the network
`build_nn` creates the network for a given list of hidden-layer sizes, e.g. `(16,)` = one layer of 16 units,
`(64, 32)` = two layers. Setting the seed first makes the starting weights reproducible.

```python
# --- Network builder ----------------------------------------------------------------------------------------------------------------
def build_nn(hidden=(16,), dropout=0.3, l2=1e-3, lr=1e-3, n_inputs=None, seed=SEED):
    keras.utils.set_random_seed(seed)                      # same seed → same starting weights
    model = keras.Sequential([keras.Input(shape=(n_inputs or X_train_np.shape[1],))])
    for units in hidden:                                   # one Dense + Dropout pair per hidden layer
        model.add(keras.layers.Dense(units, activation="relu", kernel_regularizer=keras.regularizers.l2(l2)))
        model.add(keras.layers.Dropout(dropout))
    model.add(keras.layers.Dense(1, activation="sigmoid"))  # output: podium probability
    model.compile(optimizer=keras.optimizers.Adam(lr), loss="binary_crossentropy",
                  metrics=[keras.metrics.AUC(curve="PR", name="pr_auc")])   # PR-AUC is tracked every epoch
    return model
```

### 8.6c Training with early stopping
`train_nn` trains one network for up to 300 epochs in mini-batches of 128 rows, watching validation PR-AUC after every
epoch. It returns the model (with the best epoch's weights) and its training history for the curves in 8.6f.

```python
# --- Training function -------------------------------------------------------------------------------------------------------------
def train_nn(hidden, seed, Xtr=None, ytr=None):
    Xtr = X_train_np if Xtr is None else Xtr               # default: the official training matrix
    ytr = y_train.to_numpy() if ytr is None else ytr
    model = build_nn(hidden, n_inputs=Xtr.shape[1], seed=seed)
    stop = keras.callbacks.EarlyStopping(monitor="val_pr_auc", mode="max",   # stop when validation PR-AUC
                                         patience=20, restore_best_weights=True)  # stalls for 20 epochs
    history = model.fit(Xtr, ytr, validation_data=(X_val_np, y_val.to_numpy()),
                        epochs=300, batch_size=128, callbacks=[stop], verbose=0)
    return model, history
```

### 8.6d Comparing three architectures (5 seeds each)
One small, one medium and one two-layer network, all "shallow" as the brief requires. Each is trained 5 times; the
table shows the mean and standard deviation of validation PR-AUC and how many epochs early stopping allowed.
This cell trains 15 networks and takes about 2–3 minutes.

```python
# --- Architecture comparison --------------------------------------------------------------------------------------------------------
NN_ARCHS = {"1 layer, 16 units": (16,), "1 layer, 32 units": (32,), "2 layers, 64-32": (64, 32)}

nn_rows = []
for label, hidden in NN_ARCHS.items():
    for seed in NN_SEEDS:
        model, history = train_nn(hidden, seed)
        p = model.predict(X_val_np, verbose=0).ravel()
        nn_rows.append({"architecture": label, "seed": seed, "epochs_run": len(history.history["loss"]), **val_scores(p)})
nn_seeds = pd.DataFrame(nn_rows)

# One row per architecture: mean ± std over the 5 seeds
nn_grid = (nn_seeds.groupby("architecture", sort=False)
           .agg(val_pr_auc_mean=("val_pr_auc", "mean"), val_pr_auc_std=("val_pr_auc", "std"),
                val_roc_auc_mean=("val_roc_auc", "mean"), epochs_mean=("epochs_run", "mean"))
           .sort_values("val_pr_auc_mean", ascending=False))
nn_grid.round(4)
```

**Finding:** the three architectures differ by less than their seed-to-seed spread: a bigger or deeper network does not
help on 19 tabular features. The best mean is kept, but any of the three would do.

### 8.6e Final network
The best architecture from 8.6d, retrained with the project seed (one of the 5 runs above, so it is reproducible).

**How to read the `summary()` table:** it does not show results; it lists the layers of the network and how many
numbers (weights) each layer learns.
- **Dense (first row):** the hidden layer. Each unit computes a weighted sum of the 31 input columns plus one bias,
  so it has 31 + 1 = 32 weights; **Param #** = units × 32 (e.g. 32 units → 1,024).
- **Dropout:** switches off 30 % of the hidden units at each training step. It learns nothing, so **Param #** = 0.
- **Dense (last row):** the output unit. One weight per hidden unit plus one bias (e.g. 32 + 1 = 33); the sigmoid turns
  its result into the podium probability.
- **Output Shape:** what each layer produces for one driver (`None` = any number of drivers at once):
  `(None, 32)` = one score per hidden unit, `(None, 1)` = the probability.
- **Trainable params:** the total number of weights learned (here 1,024 + 33 = 1,057); this is the size of the model.
  *Optimizer params* are Adam's internal running averages (2 per weight) and are not part of the model;
  *Total params* is the sum of the two.

The last printed line is `register`: validation PR-AUC / ROC-AUC of this network and its decision threshold chosen on validation.

```python
# --- Final neural network ------------------------------------------------------------------------------------------------------------
NN_HIDDEN = NN_ARCHS[nn_grid.index[0]]
nn, nn_history = train_nn(NN_HIDDEN, SEED)
nn.summary()
register("shallow NN", nn, nn.predict(X_train_np, verbose=0).ravel(), nn.predict(X_val_np, verbose=0).ravel())
```

### 8.6f Training curves
Loss and PR-AUC per epoch, on train and validation. The dashed line is the epoch whose weights were kept.

```python
# --- NN training curves --------------------------------------------------------------------------------------------------------------
h = nn_history.history
best_epoch = int(np.argmax(h["val_pr_auc"]))               # epoch restored by early stopping

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].plot(h["loss"], label="train")
axes[0].plot(h["val_loss"], label="validation")
axes[0].set_title("Loss (binary cross-entropy)")
axes[1].plot(h["pr_auc"], label="train")
axes[1].plot(h["val_pr_auc"], label="validation")
axes[1].set_title("PR-AUC")
for ax in axes:
    ax.axvline(best_epoch, color="grey", ls="--", lw=1)
    ax.set_xlabel("Epoch")
    ax.legend(frameon=False)
fig.suptitle(f"NN training curves ({nn_grid.index[0]}); dashed line = best validation epoch ({best_epoch}), weights restored")
plt.tight_layout()
plt.show()
```

**Reading the curves:** both losses fall quickly and flatten after ≈40 epochs. The training curves sit slightly *below*
validation in PR-AUC because Keras computes them with dropout switched on (30 % of units off), which handicaps the
network during training only. Early stopping keeps the weights from the best validation epoch (dashed line), so later
epochs, which add nothing, are discarded.

### 8.7a Are the validation differences real?
Validation has only 39 races, so small differences can be luck. A **race-level bootstrap** answers this: draw 39 races
at random *with replacement* (some races twice, some not at all), recompute every model's PR-AUC on that sample, and repeat
1,000 times. Whole races are resampled because the drivers of one race depend on each other (exactly 3 podiums).

First, the 1,000 resamples are prepared once, so every model is scored on exactly the same samples.

```python
# --- Bootstrap samples of validation races ----------------------------------------------------------------------------------------
rng = np.random.default_rng(SEED)
val_races = val_df["raceId"].to_numpy()
race_ids = np.unique(val_races)
race_rows = {r: np.flatnonzero(val_races == r) for r in race_ids}       # row positions of each race

# Each sample = the rows of 39 races drawn with replacement
boot_idx = [np.concatenate([race_rows[r] for r in rng.choice(race_ids, len(race_ids), replace=True)])
            for _ in range(1000)]
print(f"{len(boot_idx)} bootstrap samples of {len(race_ids)} races each")
```

For each model, the PR-AUC **difference to logistic regression** is computed on every sample. The middle 95 % of those
1,000 differences is the confidence interval: if it contains 0, the model is not significantly better or worse than
logistic regression. Takes ≈10 seconds.

```python
# --- PR-AUC difference vs logistic regression, with 95 % interval ----------------------------------------------------------------
yv = y_val.to_numpy()
ref = PROBS["logistic regression"]["val"]                  # the reference model

boot_rows = []
for name in PROBS:
    p = PROBS[name]["val"]
    diffs = [average_precision_score(yv[i], p[i]) - average_precision_score(yv[i], ref[i]) for i in boot_idx]
    lo, hi = np.percentile(diffs, [2.5, 97.5])             # middle 95 % of the 1,000 differences
    boot_rows.append({"model": name, "val_pr_auc": average_precision_score(yv, p),
                      "diff_vs_logreg": average_precision_score(yv, p) - average_precision_score(yv, ref),
                      "ci_low": lo, "ci_high": hi, "significant": not (lo <= 0 <= hi)})

val_compare = pd.DataFrame(boot_rows).set_index("model")
val_compare.round(3)
```

**Finding (validation):** only the grid rule is significantly worse than logistic regression (its interval is entirely
below 0). Random forest, gradient boosting and the NN are **statistically tied** with logistic regression: their
intervals contain 0.

### 8.7d A second decision rule: top 3 per race
The threshold rule ("podium if probability ≥ the validation threshold") can predict any number of podiums in a race.
But every race has **exactly 3** podiums (2.7b), so a natural alternative is: in each race, predict the podium for the
3 drivers with the highest probability. It needs no threshold at all.

```python
# --- Top-3-per-race decision rule ---------------------------------------------------------------------------------------------------
def top3_per_race(df, p):
    """1 for the 3 highest probabilities in each race, else 0."""
    rank = pd.Series(p).groupby(df["raceId"].to_numpy()).rank(ascending=False, method="first")
    return (rank.to_numpy() <= PODIUM_CUTOFF).astype(int)
```

## 8.9 Side check: does the neural network benefit from older seasons?

6.2 showed that starting training before 2003 does not help logistic regression or gradient boosting. Neural networks
usually need more data, so the same question is asked for the NN: the same architecture, trained from 1950, 1994 and 2003
(always up to 2019), scored on validation over the 5 seeds.

For each start year the preprocessing is **re-fitted on that window's own training rows** (still never on validation), so
each window gets its own medians and scaling. **This does not change the official split**; it only checks whether starting
in 2003 costs the NN anything. This cell trains 15 networks, about 3 minutes.

```python
# --- NN training-window check ------------------------------------------------------------------------------------------------------
win_rows = []
for start in [1950, 1994, 2003]:
    tr = model_df[model_df["year"].between(start, TRAIN_END)]               # training rows of this window
    pre = make_preprocessor(FEATURES).fit(tr[FEATURES])                     # preprocessing fitted on this window only
    Xtr_w = pre.transform(tr[FEATURES]).astype("float32")
    Xva_w = pre.transform(val_df[FEATURES]).astype("float32")
    for seed in NN_SEEDS:
        model = build_nn(NN_HIDDEN, n_inputs=Xtr_w.shape[1], seed=seed)
        stop = keras.callbacks.EarlyStopping(monitor="val_pr_auc", mode="max", patience=20, restore_best_weights=True)
        model.fit(Xtr_w, tr[TARGET].to_numpy(), validation_data=(Xva_w, y_val.to_numpy()),
                  epochs=300, batch_size=128, callbacks=[stop], verbose=0)
        win_rows.append({"train_start": start, "train_rows": len(tr), "seed": seed,
                         **val_scores(model.predict(Xva_w, verbose=0).ravel())})

# One row per start year: mean ± std over the 5 seeds
nn_windows = (pd.DataFrame(win_rows).groupby(["train_start", "train_rows"])
              .agg(val_pr_auc_mean=("val_pr_auc", "mean"), val_pr_auc_std=("val_pr_auc", "std"),
                   val_roc_auc_mean=("val_roc_auc", "mean")))
nn_windows.round(4)
```

**Finding:** the NN scores the same from every start year (differences no larger than the seed-to-seed spread), even with
3× more rows. More data does not help even the most data-hungry model here, which confirms the 2003 decision of 6.2 for
all model families.

## 8.10 Results, how each model works, and limitations

**Results** (numbers from our run; NN values can move by ≈0.005 between runs/CPUs):
- **Every learned model clearly beats the grid rule:** test PR-AUC ≈ 0.70–0.71 vs 0.65, ROC-AUC ≈ 0.94 vs 0.91. The
  features add real information beyond the starting position (form, standings, the car's pace in qualifying).
- **The learned models are statistically tied** (8.7a bootstrap). Gradient boosting is best on validation by a small,
  non-significant margin and is carried forward. A simple logistic regression is as good as the tree models and the NN,
  because the relationships are mostly monotone (better grid / form / standings → more podiums).
- **Test ≈ validation**, even though test is a new regulation era (2022–2024): PR-AUC drops by only ≈0.01–0.03. The
  models transfer to the new cars because the features are *relative* (positions, rates, shares), not era-specific.
- **Overfitting is visible but controlled:** the random forest and boosting score higher on train than on validation/test
  (e.g. RF train PR-AUC ≈ 0.76 vs ≈ 0.71 on test); the settings that limit tree size were chosen exactly because they
  generalise best on validation.
- **Decisions:** at the validation threshold the models find ≈72–90 % of podiums (recall) with ≈51–63 % precision. The
  top-3-per-race rule gives F1 ≈ 0.65–0.67 on test without any threshold.

| Model | How it works | Limitations |
|---|---|---|
| Grid rule | Predicts the podium for the top 3 starters. No learning. | Ignores pace, form, reliability; cannot rank drivers outside the top 3 well. |
| Logistic regression | Weighted sum of the (scaled) features → sigmoid → probability. One weight per column (8.3c plot). | Only straight-line effects on the log-odds; interactions (e.g. a penalty matters only for fast cars) must be built as features. Weights are hard to read when features are correlated (grid vs qualifying). |
| Random forest | 500 trees on bootstrap samples with random feature subsets; probability = average of the 500 trees' leaf probabilities. | Overfits with deep trees (8.4a); probabilities are not well calibrated; large and slower to explain. |
| Gradient boosting | Small trees added one at a time, each correcting the previous errors; the learning rate controls the step size. | Sensitive to settings (the default overfits, 8.5a); with only 39 validation races, the tuned settings are themselves a little noisy. |
| Shallow NN | 31 inputs → one hidden layer of 16 or 32 ReLU units (picked in 8.6d) → sigmoid; trained by gradient descent with dropout, L2 and early stopping. | Result depends on the random seed (8.6d std); a black box without XAI; no gain over simpler models on 19 tabular features. |

**General limitations:** validation has only 39 races (117 podiums), so differences below ≈0.03 PR-AUC are not
significant, and thresholds chosen there carry the same noise. Rows of the same race are not independent (exactly 3
podiums), which the models do not know; the top-3 rule is one way to add that knowledge. Unexpected events (crashes,
rain, safety cars) cannot be known before the race and cap any model's precision.

# 9. Feature-Group Ablation

**Question:** how much does each feature group contribute to the predictions?

**Method:** take the model carried forward from 8.7b (`BEST_MODEL`, gradient boosting), **remove one feature group**,
rebuild the preprocessing for the remaining features, retrain with **exactly the same settings**, and measure how much
PR-AUC drops compared with the full model. Nothing is tuned here; only the input columns change.

| Step | What |
|---|---|
| 9.1 | The 5 feature groups and the retraining helper |
| 9.2 | Drop one group at a time (plus grid and qualifying together) |
| 9.3 | The opposite view: keep only one group |
| 9.4 | Plot of the drops |
| 9.5 | Findings |

**How to read a drop:**
- **Big drop:** the group carries information that no other group has.
- **Small drop:** either the group is useless, or **another group covers for it** (e.g. `grid_pos` and `quali_pos` are
  0.97 correlated). 9.3 tells these two cases apart.

## 9.1 Setup

### 9.1a The feature groups
The 19 features from Section 6, in the 5 groups defined there (`FEATURE_GROUPS`).

```python
# --- The feature groups being ablated -----------------------------------------------------------------------------------------------
pd.DataFrame({"n_features": {g: len(c) for g, c in FEATURE_GROUPS.items()},
              "features": {g: ", ".join(c) for g, c in FEATURE_GROUPS.items()}})
```

### 9.1b Retraining helper
`run_features(features)` does the full pipeline for any list of features:
1. fits a **new preprocessor** on the training rows (`make_preprocessor` from Section 7, so the medians, scaling and missing
   indicators only cover the features that are left),
2. trains a **copy of `BEST_MODEL`** with the same hyperparameters (`clone` copies the settings, not the learned trees),
3. returns the validation and test probabilities and their PR-AUC / ROC-AUC.

The test set is used here only to *report* the drops; no choice is made on it.

```python
# --- Retrain BEST_MODEL on any subset of features ------------------------------------------------------------------------------------
from sklearn.base import clone

def run_features(features):
    """Preprocess with only `features`, retrain a copy of BEST_MODEL, return probabilities and scores."""
    _, Xtr, Xva, Xte = fit_transform_splits(features)               # new preprocessor fitted on train only
    model = clone(BEST_MODEL).fit(Xtr, y_train)                     # same hyperparameters, fresh training
    p_val, p_test = model.predict_proba(Xva)[:, 1], model.predict_proba(Xte)[:, 1]
    return {"n_features": len(features), "p_val": p_val,
            "val_pr_auc": average_precision_score(y_val, p_val), "val_roc_auc": roc_auc_score(y_val, p_val),
            "test_pr_auc": average_precision_score(y_test, p_test), "test_roc_auc": roc_auc_score(y_test, p_test)}

full = run_features(FEATURES)                                       # reference: all 19 features
print(f"{BEST_MODEL_NAME}, all {len(FEATURES)} features: val PR-AUC {full['val_pr_auc']:.3f} | "
      f"test PR-AUC {full['test_pr_auc']:.3f}")
```

The printed scores should equal the 8.7e results for gradient boosting (same features, same settings, same seed).

## 9.2 Drop one group at a time

Six runs: each of the 5 groups removed on its own, plus **grid and qualifying removed together**, because the two groups
hold almost the same information (the grid order is the qualifying order plus penalties) and may cover for each other.

Columns of the table:
- `val_drop` / `test_drop`: full-model PR-AUC **minus** the PR-AUC without the group (positive = the model got worse).
- `ci_low` / `ci_high`: 95 % interval of `val_drop` from the **same 1,000 race-level bootstrap samples as 8.7a**. If the
  interval is entirely above 0, removing the group hurts significantly.

Takes ≈10 seconds.

```python
# --- Drop-one-group ablation ---------------------------------------------------------------------------------------------------------
DROP_SETS = {f"without {g}": cols for g, cols in FEATURE_GROUPS.items()}
DROP_SETS["without grid + qualifying"] = FEATURE_GROUPS["grid"] + FEATURE_GROUPS["qualifying"]

drop_rows = []
for name, removed in DROP_SETS.items():
    r = run_features([f for f in FEATURES if f not in removed])
    # Paired bootstrap: on each resample of validation races, full PR-AUC minus reduced PR-AUC
    diffs = [average_precision_score(yv[i], full["p_val"][i]) - average_precision_score(yv[i], r["p_val"][i])
             for i in boot_idx]
    lo, hi = np.percentile(diffs, [2.5, 97.5])
    drop_rows.append({"experiment": name, "n_features": r["n_features"],
                      "val_pr_auc": r["val_pr_auc"], "val_drop": full["val_pr_auc"] - r["val_pr_auc"],
                      "ci_low": lo, "ci_high": hi, "significant": lo > 0,
                      "test_pr_auc": r["test_pr_auc"], "test_drop": full["test_pr_auc"] - r["test_pr_auc"],
                      "val_roc_auc": r["val_roc_auc"], "test_roc_auc": r["test_roc_auc"]})

ablation = pd.DataFrame(drop_rows).set_index("experiment").sort_values("val_drop", ascending=False)
ablation.round(3)
```

**Finding (drop one group):**
- **No single group is indispensable.** Removing any one of the 5 groups costs at most ≈0.017 validation PR-AUC, and every
  95 % interval contains 0. On test, removing form, qualifying, standings or history even scores slightly *higher* than the
  full model (0.718 to 0.726 vs 0.713), which is within the noise.
- **Removing grid and qualifying together is the only significant loss:** validation PR-AUC falls by 0.108 (interval 0.057
  to 0.145, entirely above 0) and test PR-AUC by 0.152, to 0.632 / 0.561. That is **below the grid rule alone** (0.648 / 0.647).
- **Why:** grid and qualifying carry the same information (where the driver starts). Remove one and the other covers for
  it; remove both and the model loses its most important signal.

## 9.3 The opposite view: keep only one group

A small drop in 9.2 can mean "useless" or "covered by another group". Training on **one group alone** separates the two:
a group that scores well on its own but caused a small drop in 9.2 is **useful but redundant**.

For reference, the grid rule from 8.2 (no learning, only the starting position) scored val PR-AUC 0.648.

```python
# --- Keep-only-one-group runs --------------------------------------------------------------------------------------------------------
only_rows = []
for g, cols in FEATURE_GROUPS.items():
    r = run_features(cols)
    only_rows.append({"experiment": f"only {g}", "n_features": r["n_features"],
                      "val_pr_auc": r["val_pr_auc"], "val_roc_auc": r["val_roc_auc"],
                      "test_pr_auc": r["test_pr_auc"], "test_roc_auc": r["test_roc_auc"]})

only_one = pd.DataFrame(only_rows).set_index("experiment").sort_values("val_pr_auc", ascending=False)
only_one.round(3)
```

**Finding (keep only one group):**
- **Qualifying alone (3 features) reaches 0.689** validation PR-AUC: better than the grid rule (0.648) and only 0.05 below
  the full model. Besides the position it includes `q1_gap_pct`, the car's pace, which the grid position does not show.
- **Standings (0.655) and form (0.625) are useful on their own**, close to the grid rule on validation, yet removing them in
  9.2 cost almost nothing. They are **useful but redundant**: both measure how competitive the driver and car currently are,
  which the starting position already largely reflects. They also transfer worse to the new era (test 0.555 and 0.520).
- **History alone is weak (0.369):** career statistics say little about the current car. Combined with the other groups it
  adds nothing measurable (9.2: −0.003).

Right: PR-AUC with **only** that group, against the full model (solid line) and the grid rule (dashed line).

## 9.5 Findings

- **The starting position is the backbone of the model.** Grid and qualifying together are the only part whose removal
  hurts significantly (−0.108 validation / −0.152 test PR-AUC); either one alone covers for the other.
- **The other groups add a small amount together, not individually.** The full model (0.740) beats the best single group
  (qualifying, 0.689) by ≈0.05, but no single group from form, standings or history makes a significant difference on its
  own, because they overlap with each other and with the grid.
- **Consequence for the model:** a smaller model (for example without the history group) would very likely perform the same.
  We keep all 19 features for comparability with Section 8, since removing them gives no significant gain either.
- **Limitation:** with 39 validation races, drops below ≈0.03 cannot be told apart from noise, so "no significant drop"
  means "no measurable contribution beyond the other groups", not "useless". Section 10 looks inside the model at the level
  of single features.

# 10. Explainability (XAI)

**Question:** *why* does the model predict what it predicts? Section 9 removed whole groups and retrained; here the trained
`BEST_MODEL` (gradient boosting) stays fixed and we look at which **single features** it uses.

| Step | Method | Scope | Question it answers |
|---|---|---|---|
| 10.1 | Setup | | A prediction function on the 19 raw features |
| 10.2 | Permutation importance | global | Which features does the model's *accuracy* depend on? |
| 10.3 | SHAP | global | Which features move the predictions most, and in which direction? |
| 10.4 | SHAP waterfall | local | Why did *this* driver get *this* probability? |
| 10.5 | LIME | local | The same question, answered by a simple model fitted around the driver |
| 10.6 | Findings | | |

**Global** = the whole model over many drivers; **local** = one prediction.

**Important:** these methods explain what the **model relies on**, not what **causes** a podium. A feature can be
important because it is correlated with the real cause (for example the car's speed), and correlated features share
their importance in ways that depend on the method.

## 10.1 Setup

### 10.1a Explaining the 19 raw features, not the 31 model columns
The model sees 31 columns (19 scaled features + 12 `missing_` flags). Explaining those would split one feature into two
pieces (`quali_pos` and `missing_quali_pos`) and show scaled values instead of real ones. So every method below works on
the **19 raw features**: `predict_raw` runs the fitted preprocessing from Section 7 and then the model, exactly as for
a new record.

```python
# --- Prediction function on the 19 raw features -------------------------------------------------------------------------------------
import shap
from lime.lime_tabular import LimeTabularExplainer

def predict_raw(X_raw):
    """Podium probability for raw feature rows: fitted preprocessing (Section 7) + BEST_MODEL."""
    X_model = pd.DataFrame(PREPROCESSOR.transform(pd.DataFrame(X_raw, columns=FEATURES)).astype(float),
                           columns=MODEL_COLUMNS)
    return BEST_MODEL.predict_proba(X_model)[:, 1]

X_test_raw = test_df[FEATURES].reset_index(drop=True)          # raw test features (with NaN where missing)
p_test_raw = predict_raw(X_test_raw)

# Sanity check: same probabilities as the test predictions of 8.7c
assert np.allclose(p_test_raw, PROBS[BEST_MODEL_NAME]["test"])
print(f"shap {shap.__version__} | explaining {BEST_MODEL_NAME} on {len(FEATURES)} raw features, "
      f"{len(X_test_raw)} test rows | test PR-AUC {average_precision_score(y_test, p_test_raw):.3f}")
```

All explanations use the **test seasons (2022–2024)**: the model never saw them during training or tuning, so this is how
it behaves on new data. Nothing is chosen or changed based on these results.

## 10.2 Permutation importance (global)

**How it works:** take one feature column in the test set and **shuffle it randomly** across rows, so it no longer
matches its driver, then score the model again. If PR-AUC falls a lot, the model depended on that feature. Shuffling
keeps the feature's values realistic (same distribution) but destroys its link to the target.

- Each feature is shuffled **10 times** with different random orders; the table shows the mean drop and its spread.
- Shuffling a raw feature also shuffles its missing-flag (the preprocessing rebuilds it), so the feature is tested as a whole.
- Takes ≈20 seconds.

*Analogy:* to see how much a team needs a player, swap them for a random person from the crowd and watch the score.

```python
# --- Permutation importance on the test set ------------------------------------------------------------------------------------------
rng_pi = np.random.default_rng(SEED)
base_pr_auc = average_precision_score(y_test, p_test_raw)
N_REPEATS = 10

perm_rows = []
for f in FEATURES:
    drops = []
    for _ in range(N_REPEATS):
        X_shuffled = X_test_raw.copy()
        X_shuffled[f] = rng_pi.permutation(X_shuffled[f].to_numpy())            # break the link between f and the target
        drops.append(base_pr_auc - average_precision_score(y_test, predict_raw(X_shuffled)))
    perm_rows.append({"feature": f, "group": next(g for g, c in FEATURE_GROUPS.items() if f in c),
                      "pr_auc_drop": np.mean(drops), "std": np.std(drops)})

perm_imp = pd.DataFrame(perm_rows).set_index("feature").sort_values("pr_auc_drop", ascending=False)
perm_imp.round(4)
```

```python
# --- Permutation importance plot ------------------------------------------------------------------------------------------------------
GROUP_COLOURS = dict(zip(FEATURE_GROUPS, PALETTE))                      # one colour per feature group
p_ = perm_imp.iloc[::-1]
fig, ax = plt.subplots(figsize=(8, 6))
ax.barh(p_.index, p_["pr_auc_drop"], xerr=p_["std"], color=p_["group"].map(GROUP_COLOURS), capsize=2)
ax.axvline(0, color="grey", lw=1)
ax.set(xlabel="Test PR-AUC lost when the feature is shuffled", title=f"Permutation importance ({BEST_MODEL_NAME})")
ax.legend(handles=[plt.Rectangle((0, 0), 1, 1, color=c) for c in GROUP_COLOURS.values()], labels=list(GROUP_COLOURS),
          frameon=False, loc="lower right", title="group")
plt.tight_layout()
plt.show()
```

**Finding (permutation importance):**
- **Two features dominate:** shuffling `grid_pos` costs 0.120 test PR-AUC and `quali_pos` 0.076. Next is the constructor's
  recent average finish (`con_avg_finish_pct_last`, 0.018). Every other feature costs less than 0.01, which is within the
  spread of the 10 shuffles.
- **Why this differs from Section 9:** there, removing the grid group cost only 0.007, because the model was **retrained**
  and switched to `quali_pos`. Here the model stays fixed: it learned to rely on `grid_pos`, so scrambling it hurts even
  though `quali_pos` holds almost the same information. Permutation importance measures what *this* model uses, the
  ablation measures what information the data needs.
- **Two negative values:** shuffling `drv_wins_before` (−0.013) or `drv_dnf_rate_last` (−0.010) makes the test score slightly
  *better*. The patterns the model learned for them in 2003–2019 do not hold in 2022–2024; with one very dominant driver,
  a high career win count no longer separates podium finishers from the rest. The effect is small but consistent across shuffles.

## 10.3 SHAP (global)

**How it works:** SHAP splits every single prediction into **one contribution per feature**:

prediction = average prediction + contribution(grid_pos) + contribution(quali_pos) + ... (one term per feature)

- A **positive** contribution pushes this driver *towards* a podium, a **negative** one *away* from it.
- The contributions are computed fairly: each feature gets its average effect over all orders in which features could be
  added to the model (the "Shapley value" from game theory). `TreeExplainer` computes this exactly for tree models.
- For gradient boosting the values are in **log-odds** (the model's internal score before the sigmoid), not in probability.

*Analogy:* splitting a restaurant bill fairly between friends according to what each one ordered.

**From 31 columns to 19 features:** SHAP values add up, so the contribution of a raw feature is simply its column's value
**plus** its `missing_` flag's value (e.g. `quali_pos` + `missing_quali_pos`).

```python
# --- SHAP values for the test set ------------------------------------------------------------------------------------------------------
explainer = shap.TreeExplainer(BEST_MODEL)
sv = explainer.shap_values(X_test)                                     # one value per row and model column
sv = sv[:, :, 1] if sv.ndim == 3 else sv                                # some shap versions return one slice per class
base_value = np.ravel(explainer.expected_value)[-1]                     # the model's average score over the training rows (log-odds)

# Sum each feature's column and its missing flag -> one value per raw feature
sv_raw = pd.DataFrame(0.0, index=range(len(X_test)), columns=FEATURES)
for j, col in enumerate(MODEL_COLUMNS):
    feature = col.replace("missing_", "", 1) if col.startswith("missing_") else col
    sv_raw[feature] += sv[:, j]

# Check: average score + contributions = the model's score for every row
logit = np.log(p_test_raw / (1 - p_test_raw))
assert np.allclose(base_value + sv_raw.sum(axis=1), logit, atol=1e-4)
print(f"base value (average log-odds score over the training rows): {base_value:.2f}")
```

### 10.3a SHAP summary plot
**How to read it:** one row per feature (most important at the top), one dot per test driver.
- **Position left/right:** that feature's contribution for that driver (right = towards podium).
- **Colour:** the feature's real value for that driver (red = high, blue = low, grey = missing).
- Example: if the `grid_pos` row has **blue dots on the right**, a *low* grid position (starting near the front) pushes
  towards a podium.

```python
# --- SHAP summary (beeswarm) plot ---------------------------------------------------------------------------------------------------
shap.summary_plot(sv_raw.to_numpy(), X_test_raw, feature_names=FEATURES, max_display=19, show=False)
plt.title(f"SHAP values on test 2022–2024 ({BEST_MODEL_NAME}, log-odds)")
plt.tight_layout()
plt.show()
```

### 10.3b SHAP and permutation importance side by side
The global SHAP importance of a feature is its **mean absolute contribution** (how much it moves predictions on average,
in either direction). Comparing the two rankings shows where the methods agree.

```python
# --- Compare the two global rankings --------------------------------------------------------------------------------------------------
global_imp = pd.DataFrame({"group": perm_imp["group"],
                           "mean_abs_shap": sv_raw.abs().mean(),
                           "permutation_drop": perm_imp["pr_auc_drop"]})
global_imp["rank_shap"] = global_imp["mean_abs_shap"].rank(ascending=False).astype(int)
global_imp["rank_permutation"] = global_imp["permutation_drop"].rank(ascending=False).astype(int)
global_imp = global_imp.sort_values("rank_shap")

# How similar are the two rankings? (1 = identical order)
from scipy.stats import spearmanr
print(f"Rank correlation SHAP vs permutation: {spearmanr(global_imp['rank_shap'], global_imp['rank_permutation'])[0]:.2f}")
global_imp.round(4)
```

**Finding (SHAP):**
- **Top features:** `quali_pos`, the constructor's recent average finish and `grid_pos`, the same top 3 as permutation
  importance. The rank correlation between the two methods is 0.69: they agree on what matters most, and differ on the
  smaller features.
- **Directions make sense in F1 terms:** a low (front) qualifying or grid position, a car that recently finished high
  (low `con_avg_finish_pct_last`), a high recent podium rate and a high championship position all push towards a podium.
- **Asymmetry:** for the few front-runners the push is large (up to ≈+1.9 log-odds for `quali_pos`), while the many midfield
  drivers get a small, similar negative push (≈−0.5). The model mostly decides *who among the front-runners* gets the podium.
- **Correlated features share credit differently by method:** SHAP ranks `quali_pos` above `grid_pos`, permutation the
  other way round; `drv_podium_rate_last` is 4th by SHAP but 9th by permutation, because other form and standings features
  carry the same information when it is shuffled.

## 10.4 SHAP for single drivers (local)

Two test drivers are explained, chosen automatically so the examples are interesting:
- **Case A, a surprise podium the model saw coming:** started 6th or further back, finished on the podium, and got the
  **highest probability** among such drivers.

```python
# --- Pick the two example drivers ----------------------------------------------------------------------------------------------------
names = clean["drivers"].set_index("driverId")[["forename", "surname"]].agg(" ".join, axis=1)
races = clean["races"].set_index("raceId")
cases = test_df[MODEL_KEYS + ["year", TARGET] + FEATURES].reset_index(drop=True).assign(p=p_test_raw)
cases["driver"] = cases["driverId"].map(names)
cases["race"] = cases["raceId"].map(races["name"]) + " " + cases["year"].astype(str)

case_a = cases[(cases["grid_pos"] >= 6) & (cases[TARGET] == 1)]["p"].idxmax()
case_b = cases[(cases["grid_pos"] <= 3) & (cases[TARGET] == 0)]["p"].idxmax()
EXAMPLES = {"A: podium from the back": case_a, "B: front-row miss": case_b}
cases.loc[list(EXAMPLES.values()), ["driver", "race", "grid_pos", "quali_pos", TARGET, "p"]].round(3)
```

### 10.4a Waterfall plots
**How to read a waterfall:** it starts at the bottom at **E[f(x)]**, the model's average score over the training rows (a typical midfield driver,
far from the podium). Each bar
is one feature's contribution for this driver: **red pushes towards a podium, blue pushes away**. The bars add up to

```python
# --- Waterfall plot for each example ---------------------------------------------------------------------------------------------------
for label, i in EXAMPLES.items():
    expl = shap.Explanation(values=sv_raw.loc[i].to_numpy(), base_values=base_value,
                            data=X_test_raw.loc[i].to_numpy(), feature_names=FEATURES)
    shap.plots.waterfall(expl, max_display=10, show=False)
    plt.title(f"{label}: {cases.loc[i, 'driver']}, {cases.loc[i, 'race']} | p(podium) = {cases.loc[i, 'p']:.2f}")
    plt.show()
```

**Finding (single drivers):**
- **Case A** (Max Verstappen, United States GP 2023, started 6th, won): probability 0.55, well above the decision threshold.
  6th on the grid still pushes upwards compared with the average driver (+0.78), and the recent podium rate of 0.9 (+0.58),
  the strong car (+0.46) and leading the championship (+0.33) add up. The model "knew" this was the dominant driver and car.
- **Case B** (Max Verstappen, Austrian GP 2024, pole position, finished 5th): probability 0.86, driven by pole in
  qualifying (+1.46) and on the grid (+1.38). The model was confidently wrong because the cause (a late-race collision) cannot
  be known before the race; this is the limit of any pre-race model.
- Both examples are the same driver: in 2022–2024 one driver dominated, so the extreme cases of the test seasons are his.

## 10.5 LIME (local)

**How it works:** to explain one driver, LIME
1. creates ≈5,000 **fake drivers** by changing that driver's feature values at random (drawn from the training data's ranges),
2. asks the model for the probability of each fake driver,
3. fits a small **weighted linear model** on these fakes, where fakes closer to the real driver count more.

The weights of that small linear model are the explanation. LIME turns each feature into a range (for example
`grid_pos <= 3.00`), so its weights read as "being in this range adds/removes this much podium probability".

*Analogy:* to understand one curvy mountain road, you only look at the 100 metres around you, where the road is almost
straight, and describe that straight piece.

**Missing values:** LIME cannot sample from missing values, so missing values are filled with the training medians
**for LIME only** (the same values the model's own imputation uses).

```python
# --- LIME explainer on the 19 raw features --------------------------------------------------------------------------------------------
train_raw = train_df[FEATURES].fillna(train_df[FEATURES].median())     # LIME needs complete rows
lime_explainer = LimeTabularExplainer(train_raw.to_numpy(), feature_names=FEATURES, class_names=["no podium", "podium"],
                                      mode="classification", discretize_continuous=True, random_state=SEED)

def lime_predict(X):
    """LIME passes numpy rows; it needs the probability of both classes."""
    p = predict_raw(pd.DataFrame(X, columns=FEATURES))
    return np.column_stack([1 - p, p])
```

### 10.5a LIME for the same two drivers
Bars to the **right (green)** push towards a podium, bars to the **left (red)** away from it.

```python
# --- LIME explanation of each example --------------------------------------------------------------------------------------------------
lime_results = {}
for label, i in EXAMPLES.items():
    row = X_test_raw.loc[i].fillna(train_df[FEATURES].median()).to_numpy()
    exp = lime_explainer.explain_instance(row, lime_predict, num_features=8, num_samples=5000)
    lime_results[label] = exp
    exp.as_pyplot_figure()
    plt.title(f"LIME, {label}: {cases.loc[i, 'driver']}, {cases.loc[i, 'race']} | p(podium) = {cases.loc[i, 'p']:.2f}")
    plt.tight_layout()
    plt.show()
```

### 10.5b Do SHAP and LIME agree?
For each example: the top 5 features by SHAP (absolute contribution) next to LIME's top 5.

```python
# --- Top-5 features: SHAP vs LIME ---------------------------------------------------------------------------------------------------------
compare_rows = []
for label, i in EXAMPLES.items():
    shap_top = sv_raw.loc[i].abs().sort_values(ascending=False).index[:5]
    lime_top = [FEATURES[j] for j, _ in sorted(lime_results[label].as_map()[1], key=lambda t: -abs(t[1]))[:5]]
    compare_rows.append({"example": label, "SHAP top 5": ", ".join(shap_top), "LIME top 5": ", ".join(lime_top),
                         "shared": len(set(shap_top) & set(lime_top))})
pd.DataFrame(compare_rows).set_index("example")
```

**Finding (LIME):**
- LIME and SHAP agree on **4 of the top 5 features** for both drivers (qualifying and grid position, recent podium rate,
  constructor's recent finish).
- **LIME's weakness is visible:** its two explanations are almost identical, although one driver started 6th and the other on
  pole. LIME cuts every feature into 4 ranges (`quali_pos <= 6.00` covers P1 to P6), so it cannot tell the two apart, and
  its weights are in probability for a straight-line model fitted to random fake drivers. SHAP uses the exact values and
  the real trees, so its local explanations are more precise.
- LIME's result also changes slightly with its random sampling; SHAP's `TreeExplainer` is exact and gives the same values
  on every run.

## 10.6 Findings

- **The model is driven by the starting position.** Every method puts qualifying and grid position first, followed by how
  strong the car currently is (`con_avg_finish_pct_last`) and the driver's recent form and championship position.
- **This matches Section 9 and F1 logic:** the starting position is the backbone; form and standings fine-tune the
  probability among front-runners and help find drivers who reach the podium from further back (case A).
- **The three methods answer different questions:** permutation importance = what the model's accuracy depends on, SHAP =
  how much and in which direction each feature moves each prediction, LIME = a simple local approximation. They agree on the
  main features and disagree on the small ones, mainly because correlated features share credit differently.
- **Limitations:** importance is **not causation** (qualifying position does not cause a podium; a fast car causes both).
  Correlated features (grid and qualifying position, form and standings) split importance in method-dependent ways. A few
  learned patterns do not transfer to the new era (negative permutation importance of `drv_wins_before`).

# 11. Inference Function

**Goal:** a function that takes **one driver-race record** (the information known after qualifying, before the race),
runs it through the **fitted preprocessing** (Section 7) and the **final model** (`BEST_MODEL`, Section 8), and returns the
podium probability and a yes/no prediction. It must also work when some values are **missing**.

| Step | What |
|---|---|
| 11.1 | `predict_podium`: from the 19 feature values to probability + label |
| 11.2 | `build_record`: the 19 features of a driver-race pair, computed from earlier races only |
| 11.3 | Demo 1: a complete record |
| 11.4 | Demo 2: partially-missing records |

## 11.1 The inference function

`predict_podium(record)` takes a dictionary `{feature name: value}`:
- Any of the 19 features may be **left out or set to `None`/`NaN`**. The preprocessing then does exactly what it did in
  training: the median from the training seasons fills the gap, and the matching `missing_` flag is set to 1 (Section 7;
  `error_on_new=False` in `make_preprocessor` was set for this case).
- Unknown keys raise an error, so a typo (e.g. `qualipos`) is not silently treated as "missing".
- The decision uses the model's **threshold chosen on validation** in Section 8 (max F1).

It returns the probability, the prediction and the list of features that were missing.

```python
# --- Inference function ------------------------------------------------------------------------------------------------------------
INFO_KEYS = {"raceId", "driverId", "driver", "race"}                 # identifiers allowed in a record, ignored by the model
THRESHOLD = THRESHOLDS[BEST_MODEL_NAME]                              # chosen on validation in Section 8

def predict_podium(record, model=BEST_MODEL, preprocessor=PREPROCESSOR, threshold=THRESHOLD):
    """Podium probability and yes/no prediction for ONE driver-race record (dict of raw feature values)."""
    unknown = set(record) - set(FEATURES) - INFO_KEYS
    if unknown:
        raise ValueError(f"Unknown fields: {sorted(unknown)}")

    # One row with all 19 features in the training order; anything not given becomes NaN (= missing)
    row = pd.DataFrame([{f: record.get(f, np.nan) for f in FEATURES}], dtype=float)
    X = pd.DataFrame(preprocessor.transform(row).astype(float), columns=MODEL_COLUMNS)   # same steps as training
    p = float(model.predict_proba(X)[0, 1])
    return {"probability": round(p, 3),
            "prediction": "podium" if p >= threshold else "no podium",
            "threshold": round(float(threshold), 3),
            "missing_features": [f for f in FEATURES if pd.isna(row.at[0, f])]}

print(f"Model: {BEST_MODEL_NAME} | decision threshold (validation, max F1): {THRESHOLD:.3f}")
```

## 11.2 Building a record from earlier races

A real record does not arrive with the 19 features ready: form, history and standings must be **computed from the races
before it**. `build_record(race_id, driver_id)` does this with the feature functions of Section 5 (`build_features`):
1. keep only races **up to and including** the target race (nothing from the future exists),
2. **erase the target race's result** (finishing position, podium), because it is not known before the race,
3. build all features and return the target driver's 19 values. Grid and qualifying come from that race itself, because
   they are known before the start.

A check confirms that the result equals the row in the modeling table (Section 5), so the inference path and the training
path produce the same features. Building the features takes a few seconds.

```python
# --- Build one record from the races before it ----------------------------------------------------------------------------------------
def build_record(race_id, driver_id):
    """The 19 features of (race_id, driver_id), computed only from information available before that race starts."""
    race_date = base_df.loc[base_df["raceId"] == race_id, "date"].iloc[0]
    history = base_df[base_df["date"] <= race_date].copy()                       # nothing after this race
    target_race = history["raceId"] == race_id
    history.loc[target_race, ["positionOrder", "position", TARGET]] = np.nan     # the result is not known yet
    feats = build_features(history, clean, FORM_WINDOW)
    row = feats[(feats["raceId"] == race_id) & (feats["driverId"] == driver_id)]
    if row.empty:
        raise ValueError(f"Driver {driver_id} did not start race {race_id}")
    return {"raceId": race_id, "driverId": driver_id, **row.iloc[0][FEATURES].to_dict()}
```

## 11.3 Demo 1: a complete record
The last race of the test seasons (Abu Dhabi 2024), for the driver who started on pole. Every feature is available.

```python
# --- Demo 1: complete record ------------------------------------------------------------------------------------------------------------
names = clean["drivers"].set_index("driverId")[["forename", "surname"]].agg(" ".join, axis=1)
races_tbl = clean["races"].set_index("raceId")

demo_race = test_df.loc[test_df["date"].idxmax(), "raceId"]                    # last race of 2024
demo_driver = test_df.loc[(test_df["raceId"] == demo_race) & (test_df["grid_pos"] == 1), "driverId"].iloc[0]
complete_record = build_record(demo_race, demo_driver)

# Same features as in the modeling table? (NaN counts as equal)
table_row = model_df.loc[(model_df["raceId"] == demo_race) & (model_df["driverId"] == demo_driver), FEATURES].iloc[0]
assert np.allclose(pd.Series(complete_record)[FEATURES].astype(float), table_row.astype(float), equal_nan=True)

actual = int(model_df.loc[table_row.name, TARGET])
print(f"{names[demo_driver]}, {races_tbl.loc[demo_race, 'name']} {races_tbl.loc[demo_race, 'year']} | "
      f"actual: {'podium' if actual else 'no podium'}")
print("Features match the modeling table ✔")
print(predict_podium(complete_record))
pd.Series(complete_record)[FEATURES].to_frame("value")
```

## 12.4 Metric choice
- **Primary: PR-AUC.** Podiums are only ≈14 % of rows. PR-AUC measures how well the podium rows are ranked at the top, and
  its random-guess value is the podium rate (≈0.15), so it shows clearly how much better than guessing a model is.
- **Secondary: ROC-AUC.** Ranking quality over all rows; easy to compare with the single-feature AUCs of Section 5, but it
  looks optimistic when positives are rare (most non-podium drivers are easy to rank low).
- **F1, precision, recall** describe the yes/no decisions at the threshold chosen on validation. The **top-3-per-race rule**
  is reported as well, because every race has exactly 3 podiums.

## 12.5 Conclusions
1. **The features add real information beyond the starting grid:** every learned model reaches test PR-AUC ≈0.70–0.71
   against 0.65 for the grid rule. The gain comes mostly from podium finishers who did *not* start in the top 3 (8.8a).
2. **A simple model is enough:** logistic regression, random forest, gradient boosting and the NN are statistically tied
   (8.7a). The relationships are mostly monotone (better start, car and form → more podiums).
3. **The starting position is the backbone** (Sections 9–10), and car strength and driver form decide among the front-runners.

## 12.6 Limitations
- **Qualifying coverage drift:** qualifying data only exists from 1994 and is incomplete until 2002 (1.3). Training therefore
  starts in 2003, where less than 2 % of the qualifying values are missing in every split. The model should not be applied
  to older seasons, where these features would mostly be imputed (11.4 shows how much predictions weaken then).
- **Era sensitivity:** the test seasons are a new regulation era. PR-AUC drops only slightly (validation 0.740 → test
  ≈0.71), but some learned patterns do not transfer: standings and form alone score clearly lower on test than on validation
  (9.3), and `drv_wins_before` has negative permutation importance on test (10.2), in seasons dominated by one driver.
- **Small validation set:** 39 races. Differences below ≈0.03 PR-AUC are not significant, and the thresholds chosen there
  carry the same uncertainty.
- **Rows are not independent:** every race has exactly 3 podiums, which the models do not know; the top-3 rule adds this
  knowledge afterwards.
- **Unpredictable events:** crashes, mechanical failures, weather and safety cars happen after the start (10.4, case B) and
  cap any pre-race model's precision.
- **Importance is not causation:** the explanations show what the model relies on, not what causes a podium.
- **Missing inputs at inference:** missing values are filled with training medians, which pull a prediction towards an
  average driver (11.4).
